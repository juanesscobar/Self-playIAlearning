````python
# AlphaZero-style Self-Play & Self-Improvement Notebook (Medical Template)
# Colab-ready, modular and didactic.
# Author: Generated for Juan Andrés — adapt and expand for portfolio/MVP.

# ============= Cell 0: Notes =============
# This notebook is written to run in Google Colab / Jupyter. It uses the
# %pip magic to install packages when executed in a notebook cell. In
# case the environment does not support notebook magics, the cell will
# fall back to pip via subprocess. This avoids SyntaxError when running
# the file in non-interactive Python executors.

# ============= Cell 1: Setup & Runtime Instructions =============
# 1) In Colab: Runtime -> Change runtime type -> GPU
# 2) Run cells in order.
# 3) This notebook contains two example environments:
#    - Connect4 (classic AlphaZero demo)
#    - ToyMedicalEnv (template for sequential medical decision problems)
# 4) Replace ToyMedicalEnv with clinical simulators or pipelines to map
#    EHR episodes into the AbstractEnv interface (see "Connect real data").

# Install dependencies (works both in notebooks and plain python)
try:
    # If running in a notebook (Colab/Jupyter), %pip magic is preferred
    get_ipython  # type: ignore
    try:
        # Use the %pip magic which ensures correct environment in notebooks
        get_ipython().run_line_magic('pip', 'install -q torch numpy tqdm')
    except Exception:
        # If magic fails for some reason, fallback to subprocess
        import subprocess, sys
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--quiet', 'torch', 'numpy', 'tqdm'])
except NameError:
    # Not running in an interactive IPython environment: use subprocess
    import subprocess, sys
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--quiet', 'torch', 'numpy', 'tqdm'])

# ============= Cell 2: Imports =============
import os
import math
import random
from collections import defaultdict, deque
from tqdm import trange, tqdm
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)

# ============= Cell 3: Utilities =============
def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)

set_seed(42)

# Helpers to save/load model
def save_model(net, path='aznet_med.pth'):
    torch.save(net.state_dict(), path)

def load_model(net, path='aznet_med.pth'):
    net.load_state_dict(torch.load(path, map_location=DEVICE))

# ============= Cell 4: Abstract Env API =============
class AbstractEnv:
    """
    Minimal interface expected by MCTS and the network:
      - clone() -> new object with identical state
      - valid_actions() -> list of valid action indices
      - step(action) -> (reward, done)
      - state_representation() -> numpy array shape (C, H, W)
      - n_actions (int) -> total action space size
    """
    pass

# ============= Cell 5: Connect4 environment =============
ROWS = 6
COLS = 7

class Connect4(AbstractEnv):
    def __init__(self):
        self.board = np.zeros((ROWS, COLS), dtype=np.int8)
        self.current_player = 1
        self.n_actions = COLS

    def clone(self):
        c = Connect4()
        c.board = self.board.copy()
        c.current_player = self.current_player
        return c

    def valid_actions(self):
        return [c for c in range(self.n_actions) if self.board[0, c] == 0]

    def step(self, action):
        for r in range(ROWS-1, -1, -1):
            if self.board[r, action] == 0:
                self.board[r, action] = self.current_player
                break
        self.current_player *= -1
        winner = self.check_winner()
        if winner is None:
            return 0.0, False
        if winner == 0:
            return 0.0, True
        return (1.0 if winner == 1 else -1.0), True

    def check_winner(self):
        b = self.board
        for r in range(ROWS):
            for c in range(COLS):
                p = b[r,c]
                if p == 0: continue
                # horiz
                if c+3 < COLS and all(b[r,c+i]==p for i in range(4)):
                    return p
                # vert
                if r+3 < ROWS and all(b[r+i,c]==p for i in range(4)):
                    return p
                # diag \
                if r+3 < ROWS and c+3 < COLS and all(b[r+i,c+i]==p for i in range(4)):
                    return p
                # diag /
                if r-3 >= 0 and c+3 < COLS and all(b[r-i,c+i]==p for i in range(4)):
                    return p
        if len(self.valid_actions()) == 0:
            return 0
        return None

    def state_representation(self):
        p = (self.board == self.current_player).astype(np.float32)
        o = (self.board == -self.current_player).astype(np.float32)
        return np.stack([p, o])

# ============= Cell 6: ToyMedicalEnv (medical template) =============
class ToyMedicalEnv(AbstractEnv):
    def __init__(self, seed=0):
        self.rng = np.random.RandomState(seed)
        # severity in [0,1], higher is worse
        self.state = np.array([self.rng.rand()*0.6 + 0.2, 0.0], dtype=np.float32)
        self.n_actions = 3
        self.t = 0
        self.max_t = 20

    def clone(self):
        c = ToyMedicalEnv()
        c.rng = np.random.RandomState()
        c.state = self.state.copy()
        c.t = self.t
        c.max_t = self.max_t
        c.n_actions = self.n_actions
        return c

    def valid_actions(self):
        return list(range(self.n_actions))

    def step(self, action):
        severity = float(self.state[0])
        noise = float(self.rng.normal(0, 0.02))
        if action == 0:
            severity = min(1.0, severity + 0.02 + noise)
        elif action == 1:
            if self.rng.rand() < 0.1:
                severity = min(1.0, severity + 0.2 + noise)
            else:
                severity = max(0.0, severity - 0.12 + noise)
        elif action == 2:
            severity = max(0.0, severity - 0.07 + noise)
        else:
            raise ValueError('Invalid action')

        self.state[0] = severity
        self.t += 1
        done = (self.t >= self.max_t) or (severity >= 0.95) or (severity <= 0.05)
        reward = -severity
        if done:
            if severity <= 0.05:
                reward = 1.0
            elif severity >= 0.95:
                reward = -1.0
            else:
                reward = 0.0
        return reward, done

    def state_representation(self):
        grid_h, grid_w = 6, 7
        severity = np.full((grid_h, grid_w), self.state[0], dtype=np.float32)
        noise = np.full((grid_h, grid_w), self.state[1], dtype=np.float32)
        return np.stack([severity, noise])

# ============= Cell 7: AZ-style Neural Network =============
class AZNet(nn.Module):
    def __init__(self, in_channels=2, action_size=7, grid_h=6, grid_w=7):
        super().__init__()
        self.grid_h = grid_h
        self.grid_w = grid_w
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU()
        )
        self.policy_conv = nn.Conv2d(64, 2, kernel_size=1)
        self.policy_fc = nn.Linear(2 * self.grid_h * self.grid_w, action_size)
        self.value_conv = nn.Conv2d(64, 1, kernel_size=1)
        self.value_fc1 = nn.Linear(self.grid_h * self.grid_w, 64)
        self.value_fc2 = nn.Linear(64, 1)

    def forward(self, state_batch):
        # Accept numpy arrays or torch tensors
        if isinstance(state_batch, np.ndarray):
            state_batch = torch.tensor(state_batch, dtype=torch.float32)
        if state_batch.dim() == 3:
            state_batch = state_batch.unsqueeze(0)
        state_batch = state_batch.to(DEVICE)
        x = self.conv(state_batch)
        p = F.relu(self.policy_conv(x))
        p = p.view(p.size(0), -1)
        p = self.policy_fc(p)

        v = F.relu(self.value_conv(x))
        v = v.view(v.size(0), -1)
        v = F.relu(self.value_fc1(v))
        v = torch.tanh(self.value_fc2(v)).squeeze(-1)
        return p, v

# ============= Cell 8: MCTS Generic =============
class MCTSNode:
    def __init__(self, prior=0.0):
        self.N = 0
        self.W = 0.0
        self.Q = 0.0
        self.P = prior
        self.children = {}

class MCTS:
    def __init__(self, net, action_size, sims=50, cpuct=1.4):
        self.net = net
        self.sims = sims
        self.cpuct = cpuct
        self.action_size = action_size

    def run(self, env):
        root = MCTSNode()
        self._expand(root, env)
        for _ in range(self.sims):
            env_copy = env.clone()
            self._simulate(env_copy, root)
        return root

    def _simulate(self, env, node):
        if not node.children:
            return self._evaluate_and_expand(env, node)

        # selection
        best_a, best_child = None, None
        best_score = -1e9
        for a, ch in node.children.items():
            u = self.cpuct * ch.P * math.sqrt(node.N + 1) / (1 + ch.N)
            score = ch.Q + u
            if score > best_score:
                best_score = score
                best_a = a
                best_child = ch
        reward, done = env.step(best_a)
        if done:
            v = reward
        else:
            v = self._simulate(env, best_child)
        # backup
        best_child.N += 1
        best_child.W += v
        best_child.Q = best_child.W / best_child.N
        node.N += 1
        node.W += v
        node.Q = node.W / node.N
        return v

    def _evaluate_and_expand(self, env, node):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
            v = float(value.cpu().numpy()[0])
        for a in env.valid_actions():
            p = float(probs[a]) if a < len(probs) else 1.0 / max(1, getattr(env, 'n_actions', len(probs)))
            node.children[a] = MCTSNode(prior=p)
        return v

    def _expand(self, node, env):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
        for a in env.valid_actions():
            p = float(probs[a]) if a < len(probs) else 1.0 / max(1, getattr(env, 'n_actions', len(probs)))
            node.children[a] = MCTSNode(prior=p)

# ============= Cell 9: Self-play dataset generation =============

def generate_self_play_data(net, env_class, n_games=16, sims=50):
    net.eval()
    dataset = []
    for _ in tqdm(range(n_games)):
        env = env_class()
        mcts = MCTS(net, action_size=getattr(env, 'n_actions', COLS), sims=sims)
        states, mcts_probs, rewards = [], [], []
        while True:
            root = mcts.run(env)
            counts = np.array([root.children[a].N if a in root.children else 0 for a in range(getattr(env, 'n_actions', COLS))], dtype=np.float32)
            if counts.sum() == 0:
                probs = np.zeros_like(counts)
                avail = env.valid_actions()
                probs[avail] = 1.0 / len(avail)
            else:
                probs = counts / (counts.sum() + 1e-8)
            states.append(env.state_representation())
            mcts_probs.append(probs)
            action = np.random.choice(np.arange(len(probs)), p=probs)
            reward, done = env.step(action)
            if done:
                for s, p in zip(states, mcts_probs):
                    dataset.append((s, p, reward))
                break
    return dataset

# ============= Cell 10: Training loop =============

def train_network(net, dataset, epochs=5, batch_size=32, lr=1e-3):
    net.train()
    opt = torch.optim.Adam(net.parameters(), lr=lr)
    for e in range(epochs):
        random.shuffle(dataset)
        losses = []
        for i in range(0, len(dataset), batch_size):
            batch = dataset[i:i+batch_size]
            states = np.stack([b[0] for b in batch])
            target_p = np.stack([b[1] for b in batch])
            target_v = np.array([b[2] for b in batch], dtype=np.float32)

            states_t = torch.tensor(states, dtype=torch.float32).to(DEVICE)
            target_p_t = torch.tensor(target_p, dtype=torch.float32).to(DEVICE)
            target_v_t = torch.tensor(target_v, dtype=torch.float32).to(DEVICE)

            logits, values = net(states_t)
            loss_v = F.mse_loss(values, target_v_t)
            loss_p = -torch.mean(torch.sum(target_p_t * F.log_softmax(logits, dim=-1), dim=1))
            loss = loss_v + loss_p

            opt.zero_grad()
            loss.backward()
            opt.step()
            losses.append(loss.item())
        print(f'Epoch {e+1}/{epochs} - loss {np.mean(losses):.4f}')

# ============= Cell 11: Example mini training =============
set_seed(123)
# Create network sized for ToyMedicalEnv (n_actions=3)
net = AZNet(in_channels=2, action_size=ToyMedicalEnv().n_actions, grid_h=6, grid_w=7).to(DEVICE)
print('Starting self-play data generation (toy, few games)')
dataset = generate_self_play_data(net, ToyMedicalEnv, n_games=8, sims=20)
print('Dataset size:', len(dataset))
if len(dataset) == 0:
    print('Warning: dataset is empty. Reduce sims or increase n_games.')
else:
    train_network(net, dataset, epochs=6, batch_size=16, lr=1e-3)
    save_model(net, 'aznet_med_colab.pth')
    print('Saved model to aznet_med_colab.pth')

# ============= Cell 12: Evaluate one episode =============

def evaluate_one(net, env_class, sims=40):
    net.eval()
    env = env_class()
    mcts = MCTS(net, action_size=getattr(env, 'n_actions', COLS), sims=sims)
    traj = []
    while True:
        root = mcts.run(env)
        counts = np.array([root.children[a].N if a in root.children else 0 for a in range(getattr(env, 'n_actions', COLS))], dtype=np.float32)
        if counts.sum() == 0:
            probs = np.zeros_like(counts)
            avail = env.valid_actions()
            probs[avail] = 1.0 / len(avail)
        else:
            probs = counts / (counts.sum() + 1e-8)
        action = probs.argmax()
        traj.append((env.state_representation(), probs, action))
        reward, done = env.step(action)
        if done:
            return reward, traj

r, traj = evaluate_one(net, ToyMedicalEnv, sims=30)
print('Episode finished, reward:', r)

# ============= Cell 13: Save / Download artifact =============
try:
    from google.colab import files
    files.download('aznet_med_colab.pth')
except Exception:
    print('Download not available in this environment. Model saved to aznet_med_colab.pth')

# ============= Cell 14: Tests (basic sanity checks) =============
# These tests are small and focused on catching runtime errors quickly.

def _test_toy_env():
    env = ToyMedicalEnv(seed=1)
    s = env.state_representation()
    assert isinstance(s, np.ndarray) and s.shape[0] == 2
    r, done = env.step(1)
    assert isinstance(r, float) and isinstance(done, bool)

def _test_connect4():
    env = Connect4()
    assert env.n_actions == 7
    assert 0 in env.valid_actions()
    r, done = env.step(0)
    assert isinstance(r, float) and isinstance(done, bool)

def run_tests():
    print('Running basic tests...')
    _test_toy_env()
    _test_connect4()
    print('All basic tests passed.')

# Run tests automatically when the notebook runs as a script
if __name__ == '__main__':
    run_tests()

# ============= Cell 15: How to connect real data / Legal considerations =============
# - For patient data (EHR): ensure consent, anonymization, and HIPAA/POPPI/GDPR compliance.
# - Process sensitive data in secure environments (VPC, private storage). Don't expose PHI in public notebooks.
# - Recommended approach: build a converter that maps EHR episodes to AbstractEnv episodes.

# ============= Cell 16: Prompt for Colab Assistant =============
PROMPT_COLAB_ASSISTANT = '''
Soy un desarrollador que está construyendo un prototipo AlphaZero-style para problemas secuenciales en medicina.
Tengo un notebook que implementa un entorno ToyMedicalEnv (estado: [severity, noise]; acciones: no-op, treatment A, treatment B).
Necesito que me ayudes a:

1) Generar 50 escenarios clínicos sintéticos variados (incluir: age, comorbidities, severidad inicial, respuesta estocástica al tratamiento) y exportarlos a CSV con campos: patient_id, age, comorbidities, initial_severity, rng_seed.

2) Proponer modificaciones realistas a ToyMedicalEnv para que incorpore: latencia del efecto del tratamiento, riesgos dependientes de edad/comorbilidad, probabilidades condicionales de efectos adversos y recompensas clínicas más finas (p. ej. QALYs aproximados).

3) Sugerir pipeline de validación offline: métricas, tests simulados, red-teaming clínico y documentación para auditoría.

4) Proponer cómo transformar escenarios en episodios para self-play: definición de episodios, terminales y diseño de recompensas.

Por favor entrega: 1) código ejemplo para generar el CSV (pandas); 2) cambios concretos al entorno; 3) lista de métricas y tests; 4) recomendaciones de seguridad y privacidad.'''

print('PROMPT listo. Copia PROMPT_COLAB_ASSISTANT y pégalo en el asistente de Colab.')
````

### Modificaciones Realistas a `ToyMedicalEnv`

Para transformar el `ToyMedicalEnv` en un simulador clínico más realista, se proponen las siguientes modificaciones conceptuales, explicando cómo integrarían los datos de los escenarios sintéticos `df_scenarios`.

#### 1. Latencia del Efecto del Tratamiento

*   **Conceptualización**: Actualmente, el efecto de una acción en `ToyMedicalEnv` es instantáneo. Una modificación realista implicaría que el impacto de un `treatment A` o `treatment B` no se ve reflejado inmediatamente en la `severity`, sino que se desarrolla a lo largo de varios pasos de tiempo. Esto imita cómo los medicamentos o intervenciones médicas tardan en hacer efecto.
*   **Cambio en `step` method**: El método `step` necesitaría un mecanismo para `track` los tratamientos administrados y sus efectos esperados a futuro. Por ejemplo, al aplicar un tratamiento, en lugar de alterar `severity` directamente, se podría añadir un 'efecto latente' o 'buffer de tratamiento' al estado del entorno. En cada paso subsiguiente, una parte de este efecto latente se aplicaría a la `severity`, y el buffer se reduciría hasta que el efecto total del tratamiento se haya manifestado. La magnitud y duración de esta latencia podrían ser parámetros del tratamiento.

#### 2. Riesgos Dependientes de Edad/Comorbilidad

*   **Conceptualización**: Las características del paciente (`age`, `comorbidities`) son cruciales en medicina. En un entorno realista, un paciente mayor o con más comorbilidades podría responder de manera diferente al tratamiento o tener una progresión de la enfermedad más agresiva.
*   **Integración de `age` y `comorbidities`**: Durante la inicialización de cada instancia de `ToyMedicalEnv`, los valores de `age` y `comorbidities` de `df_scenarios` se cargarían como atributos del objeto `env`. Luego, dentro del método `step`:
    *   **Progresión de `severity`**: La `severity` base podría aumentar más rápidamente para pacientes con mayor `age` o `comorbidities` (p. ej., un multiplicador en el aumento espontáneo de `severity`).
    *   **Efectividad del tratamiento**: La reducción de `severity` por `treatment A` o `treatment B` podría ser menor para pacientes con alta `age` o `comorbidities`.
    *   **Umbrales de `severity`**: Los umbrales de `severity` para `done` (0.05 y 0.95) o los riesgos de eventos adversos (ver siguiente punto) podrían ajustarse en función de estos factores.

#### 3. Probabilidades Condicionales de Efectos Adversos

*   **Conceptualización**: Los tratamientos médicos conllevan riesgos. Los efectos adversos no son binarios, sino que ocurren con cierta probabilidad, y esta probabilidad a menudo depende de las características del paciente y del estado actual de la enfermedad.
*   **Implementación en `step` method**: Al ejecutar una acción de tratamiento, el método `step` podría calcular una probabilidad de efecto adverso. Esta probabilidad sería una función de:
    *   `action` (ciertos tratamientos tienen más efectos adversos que otros).
    *   `age` y `comorbidities` (pacientes frágiles tienen mayor riesgo).
    *   `severity` actual (un paciente muy grave podría ser más susceptible).
    *   Si se desencadena un efecto adverso, esto podría resultar en un aumento súbito de `severity`, una penalización inmediata en la `reward`, o incluso una condición terminal temprana (`done=True`) con un `reward` muy negativo.

#### 4. Recompensas Clínicas Más Finas (p. ej., QALYs aproximados)

*   **Conceptualización**: La recompensa actual es simple (-`severity` o +/- 1.0 al final). Una recompensa más sofisticada reflejaría la calidad de vida y el tiempo vivido, como los QALYs (Quality-Adjusted Life Years).
*   **Diseño de la función de recompensa**: El `reward` en cada paso de tiempo podría ser una función de la `severity` actual, reflejando la utilidad o calidad de vida en ese momento. Por ejemplo:
    *   `reward_per_step = f(severity_t)` donde `f` es una función decreciente con la `severity` (e.g., `1 - severity_t`). Esto incentivaría mantener la `severity` baja a lo largo del tiempo.
    *   Se podría introducir un `discount factor` para reflejar la preferencia por recompensas inmediatas sobre futuras, aunque en QALYs el descuento suele ser bajo o nulo.
    *   Al final del episodio (`done=True`), se podría añadir una recompensa terminal basada en el éxito del tratamiento (`severity <= 0.05`), la supervivencia a la duración máxima (`t >= max_t`) y la `severity` final promedio, en lugar de solo los -1, 0, 1.
    *   Los eventos adversos (del punto 3) también penalizarían la recompensa de manera significativa.

#### 5. Integración de los Datos de Escenarios Sintéticos (`df_scenarios`)

La `df_scenarios` actuaría como un `pool` de perfiles de pacientes iniciales para cada simulación. Para cada juego de auto-play o evaluación:

*   Se seleccionaría una fila aleatoria (o iterativa) de `df_scenarios`.
*   Los valores de `patient_id`, `age`, `comorbidities`, `initial_severity`, y `rng_seed` de esa fila se pasarían al constructor del `ToyMedicalEnv` modificado.
*   **`patient_id`**: Serviría para identificar el escenario, útil para logging y análisis.
*   **`age` y `comorbidities`**: Se almacenarían como atributos del entorno y se usarían en el `step` method para ajustar la dinámica de `severity`, la efectividad del tratamiento y las probabilidades de efectos adversos, como se describió anteriormente.
*   **`initial_severity`**: Se usaría para establecer el estado inicial `self.state[0]` del entorno, proporcionando una variedad de puntos de partida.
*   **`rng_seed`**: Sería fundamental para asegurar la reproducibilidad de cada escenario. Cada `ToyMedicalEnv` se inicializaría con su `rng` (generador de números aleatorios) sembrado con este `rng_seed`, garantizando que, dadas las mismas acciones, el comportamiento estocástico del entorno (ruido, efectos adversos) sea idéntico para ese paciente específico.

### 1. Definir Métricas Clave

Para evaluar el rendimiento del agente AlphaZero en escenarios clínicos simulados, se proponen las siguientes métricas cuantitativas, seleccionadas por su relevancia clínica y su capacidad para informar la toma de decisiones médicas:

*   **Calidad de Vida Media (QALYs aproximados):** Los 'Quality-Adjusted Life Years' (QALYs) son una medida sintética que combina la duración y la calidad de vida. En un entorno simulado, se pueden aproximar asignando un valor de "calidad" a cada estado del paciente (inversamente proporcional a la severidad) y sumándolos a lo largo del episodio. Por ejemplo, `QALY = sum(1 - severity_t) for t in episode`. Un QALY más alto indica una mejor trayectoria de salud general.

*   **Incidencia de Eventos Adversos Graves (EAGs):** Esta métrica cuantifica la frecuencia de resultados negativos graves o efectos secundarios inducidos por el tratamiento (e.g., `severity >= 0.95`). Es crucial para evaluar la seguridad del tratamiento propuesto por el agente. En `ToyMedicalEnv`, un `reward = -1.0` al final del episodio indica un EAG.

*   **Tiempo hasta la Recuperación / Control de la Enfermedad:** Para los escenarios exitosos, se medirá el número de pasos de tiempo que tarda el agente en llevar al paciente a un estado de baja severidad. Un tiempo menor es preferible, indicando una mayor eficiencia en el tratamiento.

*   **Utilización de Recursos / Costo del Tratamiento:** Aunque `ToyMedicalEnv` no lo modela directamente, en un entorno más avanzado, esta métrica podría cuantificar el número de acciones de tratamiento 'costosas' tomadas (e.g., terapias intensivas, fármacos caros). Un buen agente debería buscar un equilibrio entre eficacia y eficiencia de recursos. Se puede simular asignando un costo a cada acción y sumándolo durante el episodio.

Estas métricas son importantes porque proporcionan una visión holística del rendimiento del agente, cubriendo la eficacia, seguridad, eficiencia y la experiencia general del paciente, aspectos críticos en la toma de decisiones clínicas.

### 2. Proponer Tipos de Tests Simulados

Para una validación robusta del agente, se deben aplicar diversas categorías de pruebas en el entorno de simulación offline, que permitan evaluar su rendimiento bajo diferentes condiciones y detectar posibles fallos o limitaciones:

    *   **Fallas en tratamientos estándar:** El agente debe ser capaz de pivotar a estrategias alternativas cuando las primeras líneas de tratamiento no son efectivas.

*   **Escenarios de Sensibilidad:** Estas pruebas implican variar sistemáticamente los parámetros del modelo del entorno para evaluar la robustez y adaptabilidad del agente. Esto ayuda a entender cómo el agente se comporta ante incertidumbres o variaciones en la realidad:
    *   **Dosis de ruido en la simulación:** Cambiar la magnitud del componente estocástico (`noise`) en la evolución de la enfermedad para ver si el agente sigue siendo efectivo con mayor o menor imprevisibilidad.
    *   **Efectividad del tratamiento:** Modificar la potencia o el perfil de riesgo-beneficio de las acciones de tratamiento para ver si el agente ajusta su política adecuadamente.
    *   **Variabilidad inter-paciente:** Evaluar cómo el agente maneja diferentes perfiles de pacientes más allá de los entrenados, para asegurar la generalización.

*   **Comparación con Líneas Base (Baselines):** Es fundamental comparar el rendimiento del agente AlphaZero con políticas de tratamiento ya establecidas o con estrategias más simples. Esto permite cuantificar el valor añadido del enfoque AlphaZero:
    *   **Guías clínicas estándar:** Simular tratamientos basados en protocolos médicos actuales.

Estas pruebas garantizan que el agente no solo funciona bien en condiciones promedio, sino que también es seguro y robusto en situaciones complejas, raras o inciertas, lo cual es vital en la medicina.

## Transformación de escenarios en episodios para self-play

### Subtask:
Explicar cómo los escenarios clínicos sintéticos se pueden transformar en episodios utilizables para el entrenamiento de self-play de AlphaZero. Esto abarcará la definición de un 'episodio', las condiciones para su terminación (terminales), y el diseño de funciones de recompensa que fomenten comportamientos deseables y penalicen los indeseables.


### Transformación de escenarios en episodios para self-play

Para el entrenamiento de self-play de AlphaZero, cada `escenario clínico sintético` generado se convierte en un `episodio` del entorno. El objetivo es que el agente aprenda a navegar estos episodios, tomando las mejores decisiones de tratamiento.

#### 1. Definición de un 'Episodio'

Un **episodio** en el contexto de `ToyMedicalEnv` (y su versión modificada) se define como una secuencia de estados, acciones, y recompensas que comienza con un paciente específico y termina cuando se alcanza una condición terminal. Cada escenario clínico sintético generado en el CSV (`synthetic_clinical_scenarios.csv`) sirve como el punto de partida para un nuevo episodio:

*   **`patient_id`**: Identificador único del paciente para el seguimiento.
*   **`age` y `comorbidities`**: Estos parámetros, aunque no están directamente en el `state` actual de `ToyMedicalEnv`, se usarían para inicializar una versión extendida del entorno, influyendo en la dinámica de la enfermedad y la respuesta al tratamiento. Por ejemplo, un paciente con mayor `age` o `comorbidities` podría tener un mayor riesgo de efectos adversos o una respuesta más lenta al tratamiento.
*   **`initial_severity`**: Este valor se mapea directamente al estado inicial `self.state[0]` de `ToyMedicalEnv`. Cada episodio comenzaría con la severidad del paciente establecida en este valor. El `rng_seed` se utilizaría para inicializar el generador de números aleatorios del entorno (`self.rng`) para ese episodio particular, asegurando que, aunque el entorno sea estocástico, las simulaciones para un paciente dado puedan ser replicadas o tener una base estocástica definida por el escenario.

Es decir, cada fila del `df_scenarios` se instanciaría como un nuevo objeto `ToyMedicalEnv(seed=rng_seed, initial_severity=initial_severity, age=age, comorbidities=comorbidities_count)` (una vez que `ToyMedicalEnv` sea extendido).

#### 2. Condiciones de Terminación (Terminales)

Las condiciones de terminación para un episodio en `ToyMedicalEnv` (o su versión mejorada) determinan cuándo un ciclo de tratamiento para un paciente ha concluido. Estas pueden ser:

*   **Éxito (Remisión)**: La severidad del paciente cae por debajo de un umbral deseado (e.g., `severity <= 0.05` en `ToyMedicalEnv`). Esto representa un resultado clínico positivo.
*   **Fallo (Deterioro Severo / Muerte)**: La severidad del paciente excede un umbral crítico (e.g., `severity >= 0.95` en `ToyMedicalEnv`). Esto indica un fracaso del tratamiento o un empeoramiento de la condición a un nivel peligroso, lo que podría implicar un evento adverso grave o la muerte.
*   **Máximo de Pasos (Tiempo Límite)**: El episodio alcanza un número predefinido de pasos o `t` (e.g., `self.t >= self.max_t` en `ToyMedicalEnv`). Esto simula un período de tratamiento limitado, después del cual, si no se ha alcanzado el éxito ni el fallo, el resultado se evalúa en función del estado final de la severidad.

Adicionalmente, en un entorno modificado, se podrían añadir condiciones de terminación relacionadas con:
*   **Eventos Adversos Irreversibles**: Si una acción causa un efecto adverso grave que no se puede revertir o que requiere una interrupción permanente del tratamiento.
*   **Estancamiento**: Si la severidad del paciente no mejora (o empeora) significativamente durante un número `X` de pasos consecutivos, indicando que el tratamiento actual no es efectivo.

#### 3. Diseño de Funciones de Recompensa

La función de recompensa es crucial para guiar al agente AlphaZero a aprender políticas clínicamente óptimas. Debería ser diseñada para reflejar el valor clínico y la calidad de vida, incentivando resultados deseables y penalizando los indeseables.

La función de recompensa propuesta para una `ToyMedicalEnv` mejorada podría incluir componentes como:

*   **Recompensa por Reducción de Severidad (`-severity` continuo)**:
    *   **Justificación Clínica**: Reducir la severidad de una enfermedad es un objetivo principal en medicina. Una penalización negativa directamente proporcional a la severidad en cada paso incentiva al agente a mantener al paciente con la menor severidad posible y a resolver la condición rápidamente.
    *   **Ejemplo**: `reward_step = -current_severity`.

*   **Recompensa Terminal por Éxito (`+1.0`)**:
    *   **Justificación Clínica**: La remisión o curación (severidad <= 0.05) es el objetivo final. Una recompensa grande y positiva al alcanzar este estado refuerza las acciones que conducen a la recuperación.
    *   **Ejemplo**: `reward_terminal_success = +1.0`.

*   **Penalización Terminal por Fallo (`-1.0`)**:
    *   **Justificación Clínica**: Un deterioro severo o la muerte (severidad >= 0.95) es el peor resultado. Una penalización grande y negativa disuade fuertemente al agente de tomar acciones que conduzcan a estos resultados.
    *   **Ejemplo**: `reward_terminal_failure = -1.0`.

*   **Recompensa Aproximada por QALYs (Quality-Adjusted Life Years)**:
    *   **Justificación Clínica**: Los QALYs miden la duración y la calidad de vida. Una forma de aproximarlos sería asignar una recompensa positiva en cada paso basada en `(1 - current_severity) * quality_factor`.
    *   **Ejemplo**: `reward_qaly = (1 - current_severity) * 0.05` (donde 0.05 podría ser el valor de un 'mes' de vida saludable).

*   **Penalización por Efectos Adversos (`-X` por evento)**:
    *   **Justificación Clínica**: Los tratamientos pueden tener efectos secundarios. Si una acción tiene una probabilidad de causar un efecto adverso, el entorno debería aplicar una penalización al `reward` en ese paso. La magnitud de `X` dependería de la gravedad del efecto adverso.
    *   **Ejemplo**: Si `action == treatment_A` y `rng.rand() < adverse_effect_prob`: `reward_adverse = -0.3`.

*   **Penalización por Tratamiento Prolongado (Coste)**:
    *   **Justificación Clínica**: En la práctica clínica, los tratamientos prolongados son más costosos y pueden generar más carga para el paciente. Una pequeña penalización negativa en cada paso (`-cost_per_step`) incentiva al agente a encontrar soluciones eficientes y rápidas.
    *   **Ejemplo**: `reward_cost = -0.01` por cada paso.

*   **Penalización por Riesgos Dependientes de Edad/Comorbilidad**: Los pacientes mayores o con más comorbilidades pueden ser más susceptibles a ciertos tratamientos. El `reward_adverse` o la dinámica de la severidad podrían ser ajustados por estos factores.
    *   **Ejemplo**: `adverse_effect_prob_age_comorbidity = base_prob + age_factor * age + comorbidity_factor * comorbidities`.

La función de recompensa final en cada paso `t` sería la suma de estas componentes. Por ejemplo:
`total_reward_at_t = (-current_severity) + (1 - current_severity)*0.05 - 0.01 + reward_adverse_event_if_any`

En el paso terminal, se sumaría la recompensa terminal (`+1.0` o `-1.0`). La combinación de estas recompensas guiará al agente a aprender una política que equilibre la reducción de la severidad, la minimización de efectos adversos y la eficiencia del tratamiento, similar a cómo un médico toma decisiones en el mundo real.

````python
import math
from collections import deque

# ============= Cell 6: Modified ToyMedicalEnv (medical template) =============
class ModifiedToyMedicalEnv(AbstractEnv):
    def __init__(self, seed=0, initial_severity=None, age=40, comorbidities=0):
        self.rng = np.random.RandomState(seed)
        self.age = age # New attribute: patient's age
        self.comorbidities = comorbidities # New attribute: number of comorbidities

        # severity in [0,1], higher is worse
        # Initial severity can be passed from scenario data, otherwise generated randomly
        self.state = np.array([initial_severity if initial_severity is not None else (self.rng.rand()*0.6 + 0.2), 0.0], dtype=np.float32)
        self.n_actions = 3 # Actions: 0=no-op, 1=treatment A, 2=treatment B
        self.t = 0
        self.max_t = 20 # Maximum number of steps in an episode

        # Latency of treatment effects: stores (action, effect_remaining_duration, total_effect_magnitude)
        # total_effect_magnitude is negative for severity reduction
        self.pending_effects = deque()

        # Constants for dynamic effects (can be tuned for realism)
        self.BASE_SEVERITY_INCREASE = 0.02 # Natural progression of severity per step
        self.TREATMENT_A_EFFECT_MAGNITUDE = 0.12 # Total severity reduction from Treatment A
        self.TREATMENT_B_EFFECT_MAGNITUDE = 0.07 # Total severity reduction from Treatment B
        self.LATENCY_DURATION_A = 3 # Steps for Treatment A effect to fully manifest
        self.LATENCY_DURATION_B = 2 # Steps for Treatment B effect to fully manifest

        # Adverse effects constants
        self.BASE_ADVERSE_PROB = 0.05 # Baseline probability of an adverse event
        self.ADVERSE_EFFECT_SEVERITY_INCREASE = 0.2 # Severity increase if an adverse event occurs
        self.ADVERSE_ACTION_MULTIPLIER = {0: 0.5, 1: 1.0, 2: 1.2} # Multiplier for adverse prob based on action

    def clone(self):
        c = ModifiedToyMedicalEnv(seed=self.rng.randint(0, 1000000), initial_severity=self.state[0], age=self.age, comorbidities=self.comorbidities)
        c.state = self.state.copy()
        c.t = self.t
        c.max_t = self.max_t
        c.n_actions = self.n_actions
        # Deep copy deque for pending effects
        c.pending_effects = deque(list(self.pending_effects))

        # Copy other constants to the cloned environment
        c.BASE_SEVERITY_INCREASE = self.BASE_SEVERITY_INCREASE
        c.TREATMENT_A_EFFECT_MAGNITUDE = self.TREATMENT_A_EFFECT_MAGNITUDE
        c.TREATMENT_B_EFFECT_MAGNITUDE = self.TREATMENT_B_EFFECT_MAGNITUDE
        c.LATENCY_DURATION_A = self.LATENCY_DURATION_A
        c.LATENCY_DURATION_B = self.LATENCY_DURATION_B
        c.BASE_ADVERSE_PROB = self.BASE_ADVERSE_PROB
        c.ADVERSE_EFFECT_SEVERITY_INCREASE = self.ADVERSE_EFFECT_SEVERITY_INCREASE
        c.ADVERSE_ACTION_MULTIPLIER = self.ADVERSE_ACTION_MULTIPLIER
        return c

    def valid_actions(self):
        return list(range(self.n_actions))

    def step(self, action):
        current_severity = float(self.state[0])
        # Stochastic noise inherent to medical processes
        noise = float(self.rng.normal(0, 0.02))
        adverse_event_penalty = 0.0
        treatment_effect_from_latent = 0.0 # Accumulator for effects from past treatments

        # --- 1. Apply pending latent effects from previous steps ---
        new_pending_effects = deque()
        for act, remaining_duration, total_magnitude in self.pending_effects:
            if remaining_duration > 0:
                # Apply a fraction of the total magnitude (assuming linear application)
                original_duration = self.LATENCY_DURATION_A if act == 1 else self.LATENCY_DURATION_B
                fractional_effect = total_magnitude / original_duration
                treatment_effect_from_latent += fractional_effect
                new_pending_effects.append((act, remaining_duration - 1, total_magnitude))
        self.pending_effects = new_pending_effects

        # --- 2. Process current action: Add new latent effects or immediate actions ---
        if action == 0: # No-op: no new latent effect, no immediate treatment effect
            pass
        elif action == 1: # Treatment A: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_A, -self.TREATMENT_A_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        elif action == 2: # Treatment B: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_B, -self.TREATMENT_B_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        else:
            raise ValueError('Invalid action')

        # --- 3. Natural progression of severity, influenced by age and comorbidities ---
        # Normalize age and comorbidities to [0,1] for scaling factors
        age_factor = (self.age - 18) / 72.0 # Assuming age range 18-90
        comorbidity_factor = self.comorbidities / 5.0 # Assuming 0-5 comorbidities

        # Base severity increase, scaled by age and comorbidities (e.g., older/more comorbidities -> faster progression)
        progression_increase = self.BASE_SEVERITY_INCREASE * (1 + age_factor * 0.5 + comorbidity_factor * 0.5)
        current_severity = min(1.0, current_severity + progression_increase + noise)

        # --- 4. Apply accumulated treatment effects (from current and latent) ---
        current_severity = max(0.0, current_severity + treatment_effect_from_latent)

        # --- 5. Conditional Adverse Effects ---
        # Probability of adverse event is conditional on action, patient factors, and current severity
        adverse_prob = self.BASE_ADVERSE_PROB
        adverse_prob *= self.ADVERSE_ACTION_MULTIPLIER.get(action, 1.0)
        adverse_prob += age_factor * 0.05 # Older patients have higher risk
        adverse_prob += comorbidity_factor * 0.05 # Patients with more comorbidities have higher risk
        adverse_prob += current_severity * 0.02 # Higher severity patients might be more fragile

        if self.rng.rand() < adverse_prob: # An adverse event occurred
            current_severity = min(1.0, current_severity + self.ADVERSE_EFFECT_SEVERITY_INCREASE)
            adverse_event_penalty = -0.5 # Immediate and significant penalty

        self.state[0] = current_severity
        self.t += 1

        # --- 6. Determine done status ---
        done = False
        if self.t >= self.max_t: # Episode ends due to time limit
            done = True
        elif current_severity >= 0.95: # Episode ends due to critical failure
            done = True
        elif current_severity <= 0.05: # Episode ends due to successful remission
            done = True

        # --- 7. Finer Clinical Rewards (QALYs approximated) ---
        # Reward components:
        # a. QALY-like reward: (1 - severity) scaled per step (higher for lower severity)
        reward_qaly_approx = (1.0 - current_severity) * 0.1 # Max 0.1 for perfect health, min 0 for max severity

        # b. Time/Cost penalty: small penalty for each step to encourage efficiency
        reward_time_cost = -0.01

        # Combine step-wise rewards
        reward_step = reward_qaly_approx + reward_time_cost + adverse_event_penalty

        # c. Terminal rewards: significant rewards/penalties for episode outcome
        if done:
            if current_severity <= 0.05: # Success
                reward_step += 1.0
            elif current_severity >= 0.95: # Failure
                reward_step -= 1.0
            # If done by max_t and not success/failure, no additional terminal bonus/penalty here.
            # Could add a continuous terminal reward based on final severity (e.g., 0.5 - current_severity)

        return reward_step, done

    def state_representation(self):
        grid_h, grid_w = 6, 7 # Keeping original grid size for CNN input compatibility

        # Create channels for different state features
        severity_channel = np.full((grid_h, grid_w), self.state[0], dtype=np.float32)
        noise_channel = np.full((grid_h, grid_w), self.state[1], dtype=np.float32)

        # Normalized age and comorbidities as static channels for the patient
        age_normalized = (self.age - 18) / 72.0 # Normalize age 18-90 to 0-1
        comorbidities_normalized = self.comorbidities / 5.0 # Normalize 0-5 comorbidities to 0-1
        age_channel = np.full((grid_h, grid_w), age_normalized, dtype=np.float32)
        comorbidities_channel = np.full((grid_h, grid_w), comorbidities_normalized, dtype=np.float32)

        # Channel for latent effects: Sum of remaining treatment effect magnitudes
        # This provides a dynamic view of pending treatment impacts
        latent_effect_sum = sum(p[2] * (p[1] / (self.LATENCY_DURATION_A if p[0] == 1 else self.LATENCY_DURATION_B)) for p in self.pending_effects) # weighted by remaining duration fraction
        latent_effect_channel = np.full((grid_h, grid_w), latent_effect_sum, dtype=np.float32)

        # Stack all channels to form the state representation
        return np.stack([severity_channel, noise_channel, age_channel, comorbidities_channel, latent_effect_channel])


# ============= Cell 7: Updated AZ-style Neural Network =============
# This AZNet is updated to accept the new number of input channels from ModifiedToyMedicalEnv
class AZNet(nn.Module):
    def __init__(self, in_channels=5, action_size=3, grid_h=6, grid_w=7): # in_channels updated to 5, action_size to 3
        super().__init__()
        self.grid_h = grid_h
        self.grid_w = grid_w
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU()
        )
        self.policy_conv = nn.Conv2d(64, 2, kernel_size=1)
        self.policy_fc = nn.Linear(2 * self.grid_h * self.grid_w, action_size)
        self.value_conv = nn.Conv2d(64, 1, kernel_size=1)
        self.value_fc1 = nn.Linear(self.grid_h * self.grid_w, 64)
        self.value_fc2 = nn.Linear(64, 1)

    def forward(self, state_batch):
        # Accept numpy arrays or torch tensors
        if isinstance(state_batch, np.ndarray):
            state_batch = torch.tensor(state_batch, dtype=torch.float32)
        # Ensure batch dimension is present
        if state_batch.dim() == 3:
            state_batch = state_batch.unsqueeze(0)
        state_batch = state_batch.to(DEVICE)

        x = self.conv(state_batch)

        p = F.relu(self.policy_conv(x))
        p = p.view(p.size(0), -1) # Flatten for FC layer
        p = self.policy_fc(p) # Policy logits

        v = F.relu(self.value_conv(x))
        v = v.view(v.size(0), -1) # Flatten for FC layer
        v = F.relu(self.value_fc1(v))
        v = torch.tanh(self.value_fc2(v)).squeeze(-1) # Value output (tanh for [-1, 1] range)
        return p, v

print("Modified ToyMedicalEnv and updated AZNet classes defined.")
````

````python
import os
import math
import random
from collections import defaultdict, deque
from tqdm import trange, tqdm
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import pandas as pd # Assuming df_scenarios is loaded as a pandas DataFrame

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)

# Helpers to set seed and save/load model
def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)

def save_model(net, path='aznet_med.pth'):
    torch.save(net.state_dict(), path)

def load_model(net, path='aznet_med.pth'):
    net.load_state_dict(torch.load(path, map_location=DEVICE))

# ============= Abstract Env API =============
class AbstractEnv:
    """
    Minimal interface expected by MCTS and the network:
      - clone() -> new object with identical state
      - valid_actions() -> list of valid action indices
      - step(action) -> (reward, done)
      - state_representation() -> numpy array shape (C, H, W)
      - n_actions (int) -> total action space size
    """
    pass

# ============= Modified ToyMedicalEnv (medical template) =============
class ModifiedToyMedicalEnv(AbstractEnv):
    def __init__(self, seed=0, initial_severity=None, age=40, comorbidities=0):
        self.rng = np.random.RandomState(seed)
        self.age = age # New attribute: patient's age
        self.comorbidities = comorbidities # New attribute: number of comorbidities

        # severity in [0,1], higher is worse
        # Initialized with initial_severity if provided, else random
        self.state = np.array([initial_severity if initial_severity is not None else (self.rng.rand()*0.6 + 0.2), 0.0], dtype=np.float32)
        self.n_actions = 3 # Actions: 0=no-op, 1=treatment A, 2=treatment B
        self.t = 0
        self.max_t = 20 # Maximum number of steps in an episode

        # Latency of treatment effects: stores (action, effect_remaining_duration, total_effect_magnitude)
        # total_effect_magnitude is negative for severity reduction
        self.pending_effects = deque()

        # Constants for dynamic effects (can be tuned for realism)
        self.BASE_SEVERITY_INCREASE = 0.02 # Natural progression of severity per step
        self.TREATMENT_A_EFFECT_MAGNITUDE = 0.12 # Total severity reduction from Treatment A
        self.TREATMENT_B_EFFECT_MAGNITUDE = 0.07 # Total severity reduction from Treatment B
        self.LATENCY_DURATION_A = 3 # Steps for Treatment A effect to fully manifest
        self.LATENCY_DURATION_B = 2 # Steps for Treatment B effect to fully manifest

        # Adverse effects constants
        self.BASE_ADVERSE_PROB = 0.05 # Baseline probability of an adverse event
        self.ADVERSE_EFFECT_SEVERITY_INCREASE = 0.2 # Severity increase if an adverse event occurs
        self.ADVERSE_ACTION_MULTIPLIER = {0: 0.5, 1: 1.0, 2: 1.2} # Multiplier for adverse prob based on action

    def clone(self):
        c = ModifiedToyMedicalEnv(seed=self.rng.randint(0, 1_000_000), initial_severity=self.state[0], age=self.age, comorbidities=self.comorbidities)
        c.state = self.state.copy()
        c.t = self.t
        c.max_t = self.max_t
        c.n_actions = self.n_actions
        # Deep copy deque for pending effects
        c.pending_effects = deque(list(self.pending_effects))

        # Copy other constants to the cloned environment
        c.BASE_SEVERITY_INCREASE = self.BASE_SEVERITY_INCREASE
        c.TREATMENT_A_EFFECT_MAGNITUDE = self.TREATMENT_A_EFFECT_MAGNITUDE
        c.TREATMENT_B_EFFECT_MAGNITUDE = self.TREATMENT_B_EFFECT_MAGNITUDE
        c.LATENCY_DURATION_A = self.LATENCY_DURATION_A
        c.LATENCY_DURATION_B = self.LATENCY_DURATION_B
        c.BASE_ADVERSE_PROB = self.BASE_ADVERSE_PROB
        c.ADVERSE_EFFECT_SEVERITY_INCREASE = self.ADVERSE_EFFECT_SEVERITY_INCREASE
        c.ADVERSE_ACTION_MULTIPLIER = self.ADVERSE_ACTION_MULTIPLIER
        return c

    def valid_actions(self):
        return list(range(self.n_actions))

    def step(self, action):
        current_severity = float(self.state[0])
        # Stochastic noise inherent to medical processes
        noise = float(self.rng.normal(0, 0.02))
        adverse_event_penalty = 0.0
        treatment_effect_from_latent = 0.0 # Accumulator for effects from past treatments

        # --- 1. Apply pending latent effects from previous steps ---
        new_pending_effects = deque()
        for act, remaining_duration, total_magnitude in self.pending_effects:
            if remaining_duration > 0:
                # Apply a fraction of the total magnitude (assuming linear application)
                original_duration = self.LATENCY_DURATION_A if act == 1 else self.LATENCY_DURATION_B
                fractional_effect = total_magnitude / original_duration
                treatment_effect_from_latent += fractional_effect
                new_pending_effects.append((act, remaining_duration - 1, total_magnitude))
        self.pending_effects = new_pending_effects

        # --- 2. Process current action: Add new latent effects or immediate actions ---
        if action == 0: # No-op: no new latent effect, no immediate treatment effect
            pass
        elif action == 1: # Treatment A: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_A, -self.TREATMENT_A_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        elif action == 2: # Treatment B: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_B, -self.TREATMENT_B_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        else:
            raise ValueError('Invalid action')

        # --- 3. Natural progression of severity, influenced by age and comorbidities ---
        # Normalize age and comorbidities to [0,1] for scaling factors
        age_factor = (self.age - 18) / 72.0 # Assuming age range 18-90
        comorbidity_factor = self.comorbidities / 5.0 # Assuming 0-5 comorbidities

        # Base severity increase, scaled by age and comorbidities (e.g., older/more comorbidities -> faster progression)
        progression_increase = self.BASE_SEVERITY_INCREASE * (1 + age_factor * 0.5 + comorbidity_factor * 0.5)
        current_severity = min(1.0, current_severity + progression_increase + noise)

        # --- 4. Apply accumulated treatment effects (from current and latent) ---
        current_severity = max(0.0, current_severity + treatment_effect_from_latent)

        # --- 5. Conditional Adverse Effects ---
        # Probability of adverse event is conditional on action, patient factors, and current severity
        adverse_prob = self.BASE_ADVERSE_PROB
        adverse_prob *= self.ADVERSE_ACTION_MULTIPLIER.get(action, 1.0)
        adverse_prob += age_factor * 0.05 # Older patients have higher risk
        adverse_prob += comorbidity_factor * 0.05 # Patients with more comorbidities have higher risk
        adverse_prob += current_severity * 0.02 # Higher severity patients might be more fragile

        if self.rng.rand() < adverse_prob: # An adverse event occurred
            current_severity = min(1.0, current_severity + self.ADVERSE_EFFECT_SEVERITY_INCREASE)
            adverse_event_penalty = -0.5 # Immediate and significant penalty

        self.state[0] = current_severity
        self.t += 1

        # --- 6. Determine done status ---
        done = False
        if self.t >= self.max_t: # Episode ends due to time limit
            done = True
        elif current_severity >= 0.95: # Episode ends due to critical failure
            done = True
        elif current_severity <= 0.05: # Episode ends due to successful remission
            done = True

        # --- 7. Finer Clinical Rewards (QALYs approximated) ---
        # Reward components:
        # a. QALY-like reward: (1 - severity) scaled per step (higher for lower severity)
        reward_qaly_approx = (1.0 - current_severity) * 0.1 # Max 0.1 for perfect health, min 0 for max severity

        # b. Time/Cost penalty: small penalty for each step to encourage efficiency
        reward_time_cost = -0.01

        # Combine step-wise rewards
        reward_step = reward_qaly_approx + reward_time_cost + adverse_event_penalty

        # c. Terminal rewards: significant rewards/penalties for episode outcome
        if done:
            if current_severity <= 0.05: # Success
                reward_step += 1.0
            elif current_severity >= 0.95: # Failure
                reward_step -= 1.0
            # If done by max_t and not success/failure, no additional terminal bonus/penalty here.
            # Could add a continuous terminal reward based on final severity (e.g., 0.5 - current_severity)

        return reward_step, done

    def state_representation(self):
        grid_h, grid_w = 6, 7 # Keeping original grid size for CNN input compatibility

        # Create channels for different state features
        severity_channel = np.full((grid_h, grid_w), self.state[0], dtype=np.float32)
        noise_channel = np.full((grid_h, grid_w), self.state[1], dtype=np.float32)

        # Normalized age and comorbidities as static channels for the patient
        age_normalized = (self.age - 18) / 72.0 # Normalize age 18-90 to 0-1
        comorbidities_normalized = self.comorbidities / 5.0 # Normalize 0-5 comorbidities to 0-1
        age_channel = np.full((grid_h, grid_w), age_normalized, dtype=np.float32)
        comorbidities_channel = np.full((grid_h, grid_w), comorbidities_normalized, dtype=np.float32)

        # Channel for latent effects: Sum of remaining treatment effect magnitudes
        # This provides a dynamic view of pending treatment impacts
        latent_effect_sum = sum(p[2] * (p[1] / (self.LATENCY_DURATION_A if p[0] == 1 else self.LATENCY_DURATION_B)) for p in self.pending_effects) # weighted by remaining duration fraction
        latent_effect_channel = np.full((grid_h, grid_w), latent_effect_sum, dtype=np.float32)

        # Stack all channels to form the state representation
        return np.stack([severity_channel, noise_channel, age_channel, comorbidities_channel, latent_effect_channel])


# ============= Updated AZ-style Neural Network =============
# This AZNet is updated to accept the new number of input channels from ModifiedToyMedicalEnv
class AZNet(nn.Module):
    def __init__(self, in_channels=5, action_size=3, grid_h=6, grid_w=7): # in_channels updated to 5, action_size to 3
        super().__init__()
        self.grid_h = grid_h
        self.grid_w = grid_w
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU()
        )
        self.policy_conv = nn.Conv2d(64, 2, kernel_size=1)
        self.policy_fc = nn.Linear(2 * self.grid_h * self.grid_w, action_size)
        self.value_conv = nn.Conv2d(64, 1, kernel_size=1)
        self.value_fc1 = nn.Linear(self.grid_h * self.grid_w, 64)
        self.value_fc2 = nn.Linear(64, 1)

    def forward(self, state_batch):
        # Accept numpy arrays or torch tensors
        if isinstance(state_batch, np.ndarray):
            state_batch = torch.tensor(state_batch, dtype=torch.float32)
        # Ensure batch dimension is present
        if state_batch.dim() == 3:
            state_batch = state_batch.unsqueeze(0)
        state_batch = state_batch.to(DEVICE)

        x = self.conv(state_batch)

        p = F.relu(self.policy_conv(x))
        p = p.view(p.size(0), -1) # Flatten for FC layer
        p = self.policy_fc(p) # Policy logits

        v = F.relu(self.value_conv(x))
        v = v.view(v.size(0), -1) # Flatten for FC layer
        v = F.relu(self.value_fc1(v))
        v = torch.tanh(self.value_fc2(v)).squeeze(-1) # Value output (tanh for [-1, 1] range)
        return p, v

# ============= MCTS Generic =============
class MCTSNode:
    def __init__(self, prior=0.0):
        self.N = 0
        self.W = 0.0
        self.Q = 0.0
        self.P = prior
        self.children = {}

class MCTS:
    def __init__(self, net, action_size, sims=50, cpuct=1.4):
        self.net = net
        self.sims = sims
        self.cpuct = cpuct
        self.action_size = action_size

    def run(self, env):
        root = MCTSNode()
        self._expand(root, env)
        for _ in range(self.sims):
            env_copy = env.clone()
            self._simulate(env_copy, root)
        return root

    def _simulate(self, env, node):
        if not node.children:
            return self._evaluate_and_expand(env, node)

        # selection
        best_a, best_child = None, None
        best_score = -1e9
        for a, ch in node.children.items():
            u = self.cpuct * ch.P * math.sqrt(node.N + 1) / (1 + ch.N)
            score = ch.Q + u
            if score > best_score:
                best_score = score
                best_a = a
                best_child = ch
        reward, done = env.step(best_a)
        if done:
            # Reward is already the terminal reward for AlphaZero's value target
            v = reward
        else:
            v = self._simulate(env, best_child)
        # backup
        best_child.N += 1
        best_child.W += v
        best_child.Q = best_child.W / best_child.N
        node.N += 1
        node.W += v
        node.Q = node.W / node.N
        return v

    def _evaluate_and_expand(self, env, node):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
            v = float(value.cpu().numpy()[0])

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability (shouldn't happen with softmax on valid actions)
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))
        return v

    def _expand(self, node, env):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))

# ============= Self-play dataset generation =============
def generate_self_play_data(net, scenarios_df, n_games=16, sims=50):
    net.eval()
    dataset = []
    scenario_indices = list(range(len(scenarios_df)))

    for i in tqdm(range(n_games), desc="Generating self-play data"):
        scenario_row = scenarios_df.iloc[scenario_indices[i % len(scenarios_df)]]

        env = ModifiedToyMedicalEnv(
            seed=int(scenario_row['rng_seed']),
            initial_severity=float(scenario_row['initial_severity']),
            age=int(scenario_row['age']),
            comorbidities=int(scenario_row['comorbidities'])
        )

        mcts = MCTS(net, action_size=env.n_actions, sims=sims)

        states, mcts_probs, rewards_history = [], [], [] # rewards_history to store step-wise rewards
        while True:
            root = mcts.run(env)

            counts = np.array([root.children[a].N if a in root.children else 0 for a in range(env.n_actions)], dtype=np.float32)
            if counts.sum() == 0:
                probs = np.zeros_like(counts)
                avail = env.valid_actions()
                if avail:
                    probs[avail] = 1.0 / len(avail)
            else:
                probs = counts / (counts.sum() + 1e-8)

            states.append(env.state_representation())
            mcts_probs.append(probs)

            if probs.sum() == 0 and env.valid_actions():
                action = random.choice(env.valid_actions())
            else:
                action = np.random.choice(np.arange(len(probs)), p=probs)

            reward, done = env.step(action)
            rewards_history.append(reward)

            if done:
                # For AlphaZero, the value target for each state in the trajectory is the final outcome of the episode.
                # Sum of rewards in ModifiedToyMedicalEnv is the episode return.
                episode_return = sum(rewards_history)
                for s, p_dist in zip(states, mcts_probs):
                    dataset.append((s, p_dist, episode_return)) # All states get the final episode return as value target
                break
    return dataset

# ============= Training loop =============
def train_network(net, dataset, epochs=5, batch_size=32, lr=1e-3):
    net.train()
    opt = torch.optim.Adam(net.parameters(), lr=lr)
    for e in range(epochs):
        random.shuffle(dataset)
        losses = []
        for i in range(0, len(dataset), batch_size):
            batch = dataset[i:i+batch_size]
            if not batch:
                continue # Skip empty batches
            states = np.stack([b[0] for b in batch])
            target_p = np.stack([b[1] for b in batch])
            target_v = np.array([b[2] for b in batch], dtype=np.float32)

            states_t = torch.tensor(states, dtype=torch.float32).to(DEVICE)
            target_p_t = torch.tensor(target_p, dtype=torch.float32).to(DEVICE)
            target_v_t = torch.tensor(target_v, dtype=torch.float32).to(DEVICE)

            logits, values = net(states_t)
            loss_v = F.mse_loss(values, target_v_t)
            loss_p = -torch.mean(torch.sum(target_p_t * F.log_softmax(logits, dim=-1), dim=1))
            loss = loss_v + loss_p

            opt.zero_grad()
            loss.backward()
            opt.step()
            losses.append(loss.item())
        print(f'Epoch {e+1}/{epochs} - loss {np.mean(losses):.4f}')

# ============= Evaluate one episode =============
def evaluate_one(net, env_callable, sims=40):
    net.eval()
    env = env_callable()
    mcts = MCTS(net, action_size=env.n_actions, sims=sims)
    traj = []
    episode_rewards = []
    while True:
        root = mcts.run(env)
        counts = np.array([root.children[a].N if a in root.children else 0 for a in range(getattr(env, 'n_actions'))], dtype=np.float32)
        if counts.sum() == 0:
            probs = np.zeros_like(counts)
            avail = env.valid_actions()
            probs[avail] = 1.0 / len(avail)
        else:
            probs = counts / (counts.sum() + 1e-8)
        action = probs.argmax()
        traj.append((env.state_representation(), probs, action))
        reward, done = env.step(action)
        episode_rewards.append(reward)
        if done:
            return sum(episode_rewards), traj


# --- Main execution logic with refined hyperparameters ---
set_seed(123)

# Load df_scenarios (assuming this was loaded in a previous cell and is available)
try:
    # This block assumes df_scenarios is globally available from a previous cell.
    # If running this cell independently, uncomment and run the following:
    # df_scenarios = pd.read_csv('synthetic_clinical_scenarios.csv')
    pass
except NameError:
    print("Error: 'df_scenarios' not found. Please ensure 'synthetic_clinical_scenarios.csv' is loaded into a DataFrame named 'df_scenarios'.")
    # Fallback for demonstration if df_scenarios isn't preloaded, create a dummy one
    data = {
    'patient_id': np.arange(1, 51),
    'age': np.random.randint(18, 91, 50),
    'comorbidities': np.random.randint(0, 6, 50),
    'initial_severity': np.random.uniform(0.2, 0.8, 50),
    'rng_seed': np.arange(50, 100)
    }
    df_scenarios = pd.DataFrame(data)
    print("Using dummy df_scenarios for demonstration.")


# Self-play data generation parameters
NEW_N_GAMES = 50 # Increased from 16 to 50
NEW_MCTS_SIMS = 75 # Adjusted from 50 to 75

# Training network parameters
NEW_EPOCHS = 10 # Increased from 6 to 10
NEW_BATCH_SIZE = 32 # Increased from 16 to 32
NEW_LR = 5e-4 # Adjusted from 1e-3 to 5e-4

print(f"New Hyperparameters:\n  n_games: {NEW_N_GAMES}\n  sims: {NEW_MCTS_SIMS}\n  epochs: {NEW_EPOCHS}\n  batch_size: {NEW_BATCH_SIZE}\n  lr: {NEW_LR}")

# Create a dummy instance of ModifiedToyMedicalEnv to get state shape and action size
dummy_env = ModifiedToyMedicalEnv(seed=0, age=50, comorbidities=1, initial_severity=0.5)
input_channels = dummy_env.state_representation().shape[0]
action_size = dummy_env.n_actions

# Initialize a new AZNet with the correct dimensions
net_refined = AZNet(in_channels=input_channels, action_size=action_size,
                    grid_h=6, grid_w=7).to(DEVICE)

print('\nStarting self-play data generation with refined parameters...')
# Call the modified generate_self_play_data function with df_scenarios and new params
dataset_refined = generate_self_play_data(net_refined, df_scenarios,
                                            n_games=NEW_N_GAMES, sims=NEW_MCTS_SIMS)
print('Refined Dataset size:', len(dataset_refined))

if len(dataset_refined) == 0:
    print('Warning: Refined dataset is empty. Reduce sims or increase n_games.')
else:
    print('\nStarting network training with refined parameters...')
    train_network(net_refined, dataset_refined, epochs=NEW_EPOCHS,
                  batch_size=NEW_BATCH_SIZE, lr=NEW_LR)
    save_model(net_refined, 'aznet_med_colab_refined.pth')
    print('Saved refined model to aznet_med_colab_refined.pth')

# Evaluate one episode with the refined environment and network
print('\nEvaluating one episode with refined network...')
# Use a different seed for evaluation to ensure varied scenarios
eval_env_args = {'age': 65, 'comorbidities': 2, 'initial_severity': 0.6, 'rng_seed': 99}
r_refined, traj_refined = evaluate_one(net_refined,
                                         lambda: ModifiedToyMedicalEnv(**eval_env_args),
                                         sims=NEW_MCTS_SIMS)
print('Episode finished with refined network, reward:', r_refined)
````

````python
import os
import math
import random
from collections import defaultdict, deque
from tqdm import trange, tqdm
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import pandas as pd # Assuming df_scenarios is loaded as a pandas DataFrame

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)

# Helpers to set seed and save/load model
def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)

def save_model(net, path='aznet_med.pth'):
    torch.save(net.state_dict(), path)

def load_model(net, path='aznet_med.pth'):
    net.load_state_dict(torch.load(path, map_location=DEVICE))

# ============= Abstract Env API =============
class AbstractEnv:
    """
    Minimal interface expected by MCTS and the network:
      - clone() -> new object with identical state
      - valid_actions() -> list of valid action indices
      - step(action) -> (reward, done)
      - state_representation() -> numpy array shape (C, H, W)
      - n_actions (int) -> total action space size
    """
    pass

# ============= Modified ToyMedicalEnv (medical template) =============
class ModifiedToyMedicalEnv(AbstractEnv):
    def __init__(self, seed=0, initial_severity=None, age=40, comorbidities=0):
        self.rng = np.random.RandomState(seed)
        self.age = age # New attribute: patient's age
        self.comorbidities = comorbidities # New attribute: number of comorbidities

        # severity in [0,1], higher is worse
        # Initialized with initial_severity if provided, else random
        self.state = np.array([initial_severity if initial_severity is not None else (self.rng.rand()*0.6 + 0.2), 0.0], dtype=np.float32)
        self.n_actions = 3 # Actions: 0=no-op, 1=treatment A, 2=treatment B
        self.t = 0
        self.max_t = 20 # Maximum number of steps in an episode

        # Latency of treatment effects: stores (action, effect_remaining_duration, total_effect_magnitude)
        # total_effect_magnitude is negative for severity reduction
        self.pending_effects = deque()

        # Constants for dynamic effects (can be tuned for realism)
        self.BASE_SEVERITY_INCREASE = 0.02 # Natural progression of severity per step
        self.TREATMENT_A_EFFECT_MAGNITUDE = 0.12 # Total severity reduction from Treatment A
        self.TREATMENT_B_EFFECT_MAGNITUDE = 0.07 # Total severity reduction from Treatment B
        self.LATENCY_DURATION_A = 3 # Steps for Treatment A effect to fully manifest
        self.LATENCY_DURATION_B = 2 # Steps for Treatment B effect to fully manifest

        # Adverse effects constants
        self.BASE_ADVERSE_PROB = 0.05 # Baseline probability of an adverse event
        self.ADVERSE_EFFECT_SEVERITY_INCREASE = 0.2 # Severity increase if an adverse event occurs
        self.ADVERSE_ACTION_MULTIPLIER = {0: 0.5, 1: 1.0, 2: 1.2} # Multiplier for adverse prob based on action

    def clone(self):
        c = ModifiedToyMedicalEnv(seed=self.rng.randint(0, 1_000_000), initial_severity=self.state[0], age=self.age, comorbidities=self.comorbidities)
        c.state = self.state.copy()
        c.t = self.t
        c.max_t = self.max_t
        c.n_actions = self.n_actions
        # Deep copy deque for pending effects
        c.pending_effects = deque(list(self.pending_effects))

        # Copy other constants to the cloned environment
        c.BASE_SEVERITY_INCREASE = self.BASE_SEVERITY_INCREASE
        c.TREATMENT_A_EFFECT_MAGNITUDE = self.TREATMENT_A_EFFECT_MAGNITUDE
        c.TREATMENT_B_EFFECT_MAGNITUDE = self.TREATMENT_B_EFFECT_MAGNITUDE
        c.LATENCY_DURATION_A = self.LATENCY_DURATION_A
        c.LATENCY_DURATION_B = self.LATENCY_DURATION_B
        c.BASE_ADVERSE_PROB = self.BASE_ADVERSE_PROB
        c.ADVERSE_EFFECT_SEVERITY_INCREASE = self.ADVERSE_EFFECT_SEVERITY_INCREASE
        c.ADVERSE_ACTION_MULTIPLIER = self.ADVERSE_ACTION_MULTIPLIER
        return c

    def valid_actions(self):
        return list(range(self.n_actions))

    def step(self, action):
        current_severity = float(self.state[0])
        # Stochastic noise inherent to medical processes
        noise = float(self.rng.normal(0, 0.02))
        adverse_event_penalty = 0.0
        treatment_effect_from_latent = 0.0 # Accumulator for effects from past treatments

        # --- 1. Apply pending latent effects from previous steps ---
        new_pending_effects = deque()
        for act, remaining_duration, total_magnitude in self.pending_effects:
            if remaining_duration > 0:
                # Apply a fraction of the total magnitude (assuming linear application)
                original_duration = self.LATENCY_DURATION_A if act == 1 else self.LATENCY_DURATION_B
                fractional_effect = total_magnitude / original_duration
                treatment_effect_from_latent += fractional_effect
                new_pending_effects.append((act, remaining_duration - 1, total_magnitude))
        self.pending_effects = new_pending_effects

        # --- 2. Process current action: Add new latent effects or immediate actions ---
        if action == 0: # No-op: no new latent effect, no immediate treatment effect
            pass
        elif action == 1: # Treatment A: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_A, -self.TREATMENT_A_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        elif action == 2: # Treatment B: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_B, -self.TREATMENT_B_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        else:
            raise ValueError('Invalid action')

        # --- 3. Natural progression of severity, influenced by age and comorbidities ---
        # Normalize age and comorbidities to [0,1] for scaling factors
        age_factor = (self.age - 18) / 72.0 # Assuming age range 18-90
        comorbidity_factor = self.comorbidities / 5.0 # Assuming 0-5 comorbidities

        # Base severity increase, scaled by age and comorbidities (e.g., older/more comorbidities -> faster progression)
        progression_increase = self.BASE_SEVERITY_INCREASE * (1 + age_factor * 0.5 + comorbidity_factor * 0.5)
        current_severity = min(1.0, current_severity + progression_increase + noise)

        # --- 4. Apply accumulated treatment effects (from current and latent) ---
        current_severity = max(0.0, current_severity + treatment_effect_from_latent)

        # --- 5. Conditional Adverse Effects ---
        # Probability of adverse event is conditional on action, patient factors, and current severity
        adverse_prob = self.BASE_ADVERSE_PROB
        adverse_prob *= self.ADVERSE_ACTION_MULTIPLIER.get(action, 1.0)
        adverse_prob += age_factor * 0.05 # Older patients have higher risk
        adverse_prob += comorbidity_factor * 0.05 # Patients with more comorbidities have higher risk
        adverse_prob += current_severity * 0.02 # Higher severity patients might be more fragile

        if self.rng.rand() < adverse_prob: # An adverse event occurred
            current_severity = min(1.0, current_severity + self.ADVERSE_EFFECT_SEVERITY_INCREASE)
            adverse_event_penalty = -0.5 # Immediate and significant penalty

        self.state[0] = current_severity
        self.t += 1

        # --- 6. Determine done status ---
        done = False
        if self.t >= self.max_t: # Episode ends due to time limit
            done = True
        elif current_severity >= 0.95: # Episode ends due to critical failure
            done = True
        elif current_severity <= 0.05: # Episode ends due to successful remission
            done = True

        # --- 7. Finer Clinical Rewards (QALYs approximated) ---
        # Reward components:
        # a. QALY-like reward: (1 - severity) scaled per step (higher for lower severity)
        reward_qaly_approx = (1.0 - current_severity) * 0.1 # Max 0.1 for perfect health, min 0 for max severity

        # b. Time/Cost penalty: small penalty for each step to encourage efficiency
        reward_time_cost = -0.01

        # Combine step-wise rewards
        reward_step = reward_qaly_approx + reward_time_cost + adverse_event_penalty

        # c. Terminal rewards: significant rewards/penalties for episode outcome
        if done:
            if current_severity <= 0.05: # Success
                reward_step += 1.0
            elif current_severity >= 0.95: # Failure
                reward_step -= 1.0
            # If done by max_t and not success/failure, no additional terminal bonus/penalty here.
            # Could add a continuous terminal reward based on final severity (e.g., 0.5 - current_severity)

        return reward_step, done

    def state_representation(self):
        grid_h, grid_w = 6, 7 # Keeping original grid size for CNN input compatibility

        # Create channels for different state features
        severity_channel = np.full((grid_h, grid_w), self.state[0], dtype=np.float32)
        noise_channel = np.full((grid_h, grid_w), self.state[1], dtype=np.float32)

        # Normalized age and comorbidities as static channels for the patient
        age_normalized = (self.age - 18) / 72.0 # Normalize age 18-90 to 0-1
        comorbidities_normalized = self.comorbidities / 5.0 # Normalize 0-5 comorbidities to 0-1
        age_channel = np.full((grid_h, grid_w), age_normalized, dtype=np.float32)
        comorbidities_channel = np.full((grid_h, grid_w), comorbidities_normalized, dtype=np.float32)

        # Channel for latent effects: Sum of remaining treatment effect magnitudes
        # This provides a dynamic view of pending treatment impacts
        latent_effect_sum = sum(p[2] * (p[1] / (self.LATENCY_DURATION_A if p[0] == 1 else self.LATENCY_DURATION_B)) for p in self.pending_effects) # weighted by remaining duration fraction
        latent_effect_channel = np.full((grid_h, grid_w), latent_effect_sum, dtype=np.float32)

        # Stack all channels to form the state representation
        return np.stack([severity_channel, noise_channel, age_channel, comorbidities_channel, latent_effect_channel])


# ============= Updated AZ-style Neural Network =============
# This AZNet is updated to accept the new number of input channels from ModifiedToyMedicalEnv
class AZNet(nn.Module):
    def __init__(self, in_channels=5, action_size=3, grid_h=6, grid_w=7): # in_channels updated to 5, action_size to 3
        super().__init__()
        self.grid_h = grid_h
        self.grid_w = grid_w
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU()
        )
        self.policy_conv = nn.Conv2d(64, 2, kernel_size=1)
        self.policy_fc = nn.Linear(2 * self.grid_h * self.grid_w, action_size)
        self.value_conv = nn.Conv2d(64, 1, kernel_size=1)
        self.value_fc1 = nn.Linear(self.grid_h * self.grid_w, 64)
        self.value_fc2 = nn.Linear(64, 1)

    def forward(self, state_batch):
        # Accept numpy arrays or torch tensors
        if isinstance(state_batch, np.ndarray):
            state_batch = torch.tensor(state_batch, dtype=torch.float32)
        # Ensure batch dimension is present
        if state_batch.dim() == 3:
            state_batch = state_batch.unsqueeze(0)
        state_batch = state_batch.to(DEVICE)

        x = self.conv(state_batch)

        p = F.relu(self.policy_conv(x))
        p = p.view(p.size(0), -1) # Flatten for FC layer
        p = self.policy_fc(p) # Policy logits

        v = F.relu(self.value_conv(x))
        v = v.view(v.size(0), -1) # Flatten for FC layer
        v = F.relu(self.value_fc1(v))
        v = torch.tanh(self.value_fc2(v)).squeeze(-1) # Value output (tanh for [-1, 1] range)
        return p, v

# ============= MCTS Generic =============
class MCTSNode:
    def __init__(self, prior=0.0):
        self.N = 0
        self.W = 0.0
        self.Q = 0.0
        self.P = prior
        self.children = {}

class MCTS:
    def __init__(self, net, action_size, sims=50, cpuct=1.4):
        self.net = net
        self.sims = sims
        self.cpuct = cpuct
        self.action_size = action_size

    def run(self, env):
        root = MCTSNode()
        self._expand(root, env)
        for _ in range(self.sims):
            env_copy = env.clone()
            self._simulate(env_copy, root)
        return root

    def _simulate(self, env, node):
        if not node.children:
            return self._evaluate_and_expand(env, node)

        # selection
        best_a, best_child = None, None
        best_score = -1e9
        for a, ch in node.children.items():
            u = self.cpuct * ch.P * math.sqrt(node.N + 1) / (1 + ch.N)
            score = ch.Q + u
            if score > best_score:
                best_score = score
                best_a = a
                best_child = ch
        reward, done = env.step(best_a)
        if done:
            # Reward is already the terminal reward for AlphaZero's value target
            v = reward
        else:
            v = self._simulate(env, best_child)
        # backup
        best_child.N += 1
        best_child.W += v
        best_child.Q = best_child.W / best_child.N
        node.N += 1
        node.W += v
        node.Q = node.W / node.N
        return v

    def _evaluate_and_expand(self, env, node):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
            v = float(value.cpu().numpy()[0])

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability (shouldn't happen with softmax on valid actions)
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))
        return v

    def _expand(self, node, env):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))

# ============= Self-play dataset generation =============
def generate_self_play_data(net, scenarios_df, n_games=16, sims=50):
    net.eval()
    dataset = []
    scenario_indices = list(range(len(scenarios_df)))

    for i in tqdm(range(n_games), desc="Generating self-play data"):
        scenario_row = scenarios_df.iloc[scenario_indices[i % len(scenarios_df)]]

        env = ModifiedToyMedicalEnv(
            seed=int(scenario_row['rng_seed']),
            initial_severity=float(scenario_row['initial_severity']),
            age=int(scenario_row['age']),
            comorbidities=int(scenario_row['comorbidities'])
        )

        mcts = MCTS(net, action_size=env.n_actions, sims=sims)

        states, mcts_probs, rewards_history = [], [], [] # rewards_history to store step-wise rewards
        while True:
            root = mcts.run(env)

            counts = np.array([root.children[a].N if a in root.children else 0 for a in range(env.n_actions)], dtype=np.float32)
            if counts.sum() == 0:
                probs = np.zeros_like(counts)
                avail = env.valid_actions()
                if avail:
                    probs[avail] = 1.0 / len(avail)
            else:
                probs = counts / (counts.sum() + 1e-8)

            states.append(env.state_representation())
            mcts_probs.append(probs)

            if probs.sum() == 0 and env.valid_actions():
                action = random.choice(env.valid_actions())
            else:
                action = np.random.choice(np.arange(len(probs)), p=probs)

            reward, done = env.step(action)
            rewards_history.append(reward)

            if done:
                # For AlphaZero, the value target for each state in the trajectory is the final outcome of the episode.
                # Sum of rewards in ModifiedToyMedicalEnv is the episode return.
                episode_return = sum(rewards_history)
                for s, p_dist in zip(states, mcts_probs):
                    dataset.append((s, p_dist, episode_return)) # All states get the final episode return as value target
                break
    return dataset

# ============= Training loop =============
def train_network(net, dataset, epochs=5, batch_size=32, lr=1e-3):
    net.train()
    opt = torch.optim.Adam(net.parameters(), lr=lr)
    for e in range(epochs):
        random.shuffle(dataset)
        losses = []
        for i in range(0, len(dataset), batch_size):
            batch = dataset[i:i+batch_size]
            if not batch:
                continue # Skip empty batches
            states = np.stack([b[0] for b in batch])
            target_p = np.stack([b[1] for b in batch])
            target_v = np.array([b[2] for b in batch], dtype=np.float32)

            states_t = torch.tensor(states, dtype=torch.float32).to(DEVICE)
            target_p_t = torch.tensor(target_p, dtype=torch.float32).to(DEVICE)
            target_v_t = torch.tensor(target_v, dtype=torch.float32).to(DEVICE)

            logits, values = net(states_t)
            loss_v = F.mse_loss(values, target_v_t)
            loss_p = -torch.mean(torch.sum(target_p_t * F.log_softmax(logits, dim=-1), dim=1))
            loss = loss_v + loss_p

            opt.zero_grad()
            loss.backward()
            opt.step()
            losses.append(loss.item())
        print(f'Epoch {e+1}/{epochs} - loss {np.mean(losses):.4f}')

# ============= Evaluate one episode =============
def evaluate_one(net, env_callable, sims=40):
    net.eval()
    env = env_callable()
    mcts = MCTS(net, action_size=env.n_actions, sims=sims)
    traj = []
    episode_rewards = []
    while True:
        root = mcts.run(env)
        counts = np.array([root.children[a].N if a in root.children else 0 for a in range(getattr(env, 'n_actions'))], dtype=np.float32)
        if counts.sum() == 0:
            probs = np.zeros_like(counts)
            avail = env.valid_actions()
            probs[avail] = 1.0 / len(avail)
        else:
            probs = counts / (counts.sum() + 1e-8)
        action = probs.argmax()
        traj.append((env.state_representation(), probs, action))
        reward, done = env.step(action)
        episode_rewards.append(reward)
        if done:
            return sum(episode_rewards), traj


# --- Main execution logic with refined hyperparameters ---
set_seed(123)

# Explicitly load df_scenarios here to ensure it's defined.
# This assumes 'synthetic_clinical_scenarios.csv' exists in the current environment.
try:
    df_scenarios = pd.read_csv('synthetic_clinical_scenarios.csv')
    print(f"Loaded {len(df_scenarios)} synthetic clinical scenarios from 'synthetic_clinical_scenarios.csv'.")
except FileNotFoundError:
    print("Error: 'synthetic_clinical_scenarios.csv' not found. Creating a dummy DataFrame for demonstration.")
    # Fallback if df_scenarios isn't preloaded and file not found, create a dummy one
    data = {
    'patient_id': np.arange(1, 51),
    'age': np.random.randint(18, 91, 50),
    'comorbidities': np.random.randint(0, 6, 50),
    'initial_severity': np.random.uniform(0.2, 0.8, 50),
    'rng_seed': np.arange(50, 100)
    }
    df_scenarios = pd.DataFrame(data)
    print("Using dummy df_scenarios for demonstration.")


# Self-play data generation parameters
NEW_N_GAMES = 50 # Increased from 16 to 50
NEW_MCTS_SIMS = 75 # Adjusted from 50 to 75

# Training network parameters
NEW_EPOCHS = 10 # Increased from 6 to 10
NEW_BATCH_SIZE = 32 # Increased from 16 to 32
NEW_LR = 5e-4 # Adjusted from 1e-3 to 5e-4

print(f"New Hyperparameters:\n  n_games: {NEW_N_GAMES}\n  sims: {NEW_MCTS_SIMS}\n  epochs: {NEW_EPOCHS}\n  batch_size: {NEW_BATCH_SIZE}\n  lr: {NEW_LR}")

# Create a dummy instance of ModifiedToyMedicalEnv to get state shape and action size
dummy_env = ModifiedToyMedicalEnv(seed=0, age=50, comorbidities=1, initial_severity=0.5)
input_channels = dummy_env.state_representation().shape[0]
action_size = dummy_env.n_actions

# Initialize a new AZNet with the correct dimensions
net_refined = AZNet(in_channels=input_channels, action_size=action_size,
                    grid_h=6, grid_w=7).to(DEVICE)

print('\nStarting self-play data generation with refined parameters...')
# Call the modified generate_self_play_data function with df_scenarios and new params
dataset_refined = generate_self_play_data(net_refined, df_scenarios,
                                            n_games=NEW_N_GAMES, sims=NEW_MCTS_SIMS)
print('Refined Dataset size:', len(dataset_refined))

if len(dataset_refined) == 0:
    print('Warning: Refined dataset is empty. Reduce sims or increase n_games.')
else:
    print('\nStarting network training with refined parameters...')
    train_network(net_refined, dataset_refined, epochs=NEW_EPOCHS,
                  batch_size=NEW_BATCH_SIZE, lr=NEW_LR)
    save_model(net_refined, 'aznet_med_colab_refined.pth')
    print('Saved refined model to aznet_med_colab_refined.pth')

# Evaluate one episode with the refined environment and network
print('\nEvaluating one episode with refined network...')
# Use a different seed for evaluation to ensure varied scenarios
eval_env_args = {'age': 65, 'comorbidities': 2, 'initial_severity': 0.6, 'rng_seed': 99}
r_refined, traj_refined = evaluate_one(net_refined,
                                         lambda: ModifiedToyMedicalEnv(**eval_env_args),
                                         sims=NEW_MCTS_SIMS)
print('Episode finished with refined network, reward:', r_refined)
````

````python
import os
import math
import random
from collections import defaultdict, deque
from tqdm import trange, tqdm
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import pandas as pd # Assuming df_scenarios is loaded as a pandas DataFrame

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', DEVICE)

# Helpers to set seed and save/load model
def set_seed(s=42):
    random.seed(s)
    np.random.seed(s)
    torch.manual_seed(s)

def save_model(net, path='aznet_med.pth'):
    torch.save(net.state_dict(), path)

def load_model(net, path='aznet_med.pth'):
    net.load_state_dict(torch.load(path, map_location=DEVICE))

# ============= Abstract Env API =============
class AbstractEnv:
    """
    Minimal interface expected by MCTS and the network:
      - clone() -> new object with identical state
      - valid_actions() -> list of valid action indices
      - step(action) -> (reward, done)
      - state_representation() -> numpy array shape (C, H, W)
      - n_actions (int) -> total action space size
    """
    pass

# ============= Modified ToyMedicalEnv (medical template) =============
class ModifiedToyMedicalEnv(AbstractEnv):
    def __init__(self, seed=0, initial_severity=None, age=40, comorbidities=0):
        self.rng = np.random.RandomState(seed)
        self.age = age # New attribute: patient's age
        self.comorbidities = comorbidities # New attribute: number of comorbidities

        # severity in [0,1], higher is worse
        # Initialized with initial_severity if provided, else random
        self.state = np.array([initial_severity if initial_severity is not None else (self.rng.rand()*0.6 + 0.2), 0.0], dtype=np.float32)
        self.n_actions = 3 # Actions: 0=no-op, 1=treatment A, 2=treatment B
        self.t = 0
        self.max_t = 20 # Maximum number of steps in an episode

        # Latency of treatment effects: stores (action, effect_remaining_duration, total_effect_magnitude)
        # total_effect_magnitude is negative for severity reduction
        self.pending_effects = deque()

        # Constants for dynamic effects (can be tuned for realism)
        self.BASE_SEVERITY_INCREASE = 0.02 # Natural progression of severity per step
        self.TREATMENT_A_EFFECT_MAGNITUDE = 0.12 # Total severity reduction from Treatment A
        self.TREATMENT_B_EFFECT_MAGNITUDE = 0.07 # Total severity reduction from Treatment B
        self.LATENCY_DURATION_A = 3 # Steps for Treatment A effect to fully manifest
        self.LATENCY_DURATION_B = 2 # Steps for Treatment B effect to fully manifest

        # Adverse effects constants
        self.BASE_ADVERSE_PROB = 0.05 # Baseline probability of an adverse event
        self.ADVERSE_EFFECT_SEVERITY_INCREASE = 0.2 # Severity increase if an adverse event occurs
        self.ADVERSE_ACTION_MULTIPLIER = {0: 0.5, 1: 1.0, 2: 1.2} # Multiplier for adverse prob based on action

    def clone(self):
        c = ModifiedToyMedicalEnv(seed=self.rng.randint(0, 1_000_000), initial_severity=self.state[0], age=self.age, comorbidities=self.comorbidities)
        c.state = self.state.copy()
        c.t = self.t
        c.max_t = self.max_t
        c.n_actions = self.n_actions
        # Deep copy deque for pending effects
        c.pending_effects = deque(list(self.pending_effects))

        # Copy other constants to the cloned environment
        c.BASE_SEVERITY_INCREASE = self.BASE_SEVERITY_INCREASE
        c.TREATMENT_A_EFFECT_MAGNITUDE = self.TREATMENT_A_EFFECT_MAGNITUDE
        c.TREATMENT_B_EFFECT_MAGNITUDE = self.TREATMENT_B_EFFECT_MAGNITUDE
        c.LATENCY_DURATION_A = self.LATENCY_DURATION_A
        c.LATENCY_DURATION_B = self.LATENCY_DURATION_B
        c.BASE_ADVERSE_PROB = self.BASE_ADVERSE_PROB
        c.ADVERSE_EFFECT_SEVERITY_INCREASE = self.ADVERSE_EFFECT_SEVERITY_INCREASE
        c.ADVERSE_ACTION_MULTIPLIER = self.ADVERSE_ACTION_MULTIPLIER
        return c

    def valid_actions(self):
        return list(range(self.n_actions))

    def step(self, action):
        current_severity = float(self.state[0])
        # Stochastic noise inherent to medical processes
        noise = float(self.rng.normal(0, 0.02))
        adverse_event_penalty = 0.0
        treatment_effect_from_latent = 0.0 # Accumulator for effects from past treatments

        # --- 1. Apply pending latent effects from previous steps ---
        new_pending_effects = deque()
        for act, remaining_duration, total_magnitude in self.pending_effects:
            if remaining_duration > 0:
                # Apply a fraction of the total magnitude (assuming linear application)
                original_duration = self.LATENCY_DURATION_A if act == 1 else self.LATENCY_DURATION_B
                fractional_effect = total_magnitude / original_duration
                treatment_effect_from_latent += fractional_effect
                new_pending_effects.append((act, remaining_duration - 1, total_magnitude))
        self.pending_effects = new_pending_effects

        # --- 2. Process current action: Add new latent effects or immediate actions ---
        if action == 0: # No-op: no new latent effect, no immediate treatment effect
            pass
        elif action == 1: # Treatment A: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_A, -self.TREATMENT_A_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        elif action == 2: # Treatment B: add latent effect
            self.pending_effects.append((action, self.LATENCY_DURATION_B, -self.TREATMENT_B_EFFECT_MAGNITUDE)) # negative magnitude for severity reduction
        else:
            raise ValueError('Invalid action')

        # --- 3. Natural progression of severity, influenced by age and comorbidities ---
        # Normalize age and comorbidities to [0,1] for scaling factors
        age_factor = (self.age - 18) / 72.0 # Assuming age range 18-90
        comorbidity_factor = self.comorbidities / 5.0 # Assuming 0-5 comorbidities

        # Base severity increase, scaled by age and comorbidities (e.g., older/more comorbidities -> faster progression)
        progression_increase = self.BASE_SEVERITY_INCREASE * (1 + age_factor * 0.5 + comorbidity_factor * 0.5)
        current_severity = min(1.0, current_severity + progression_increase + noise)

        # --- 4. Apply accumulated treatment effects (from current and latent) ---
        current_severity = max(0.0, current_severity + treatment_effect_from_latent)

        # --- 5. Conditional Adverse Effects ---
        # Probability of adverse event is conditional on action, patient factors, and current severity
        adverse_prob = self.BASE_ADVERSE_PROB
        adverse_prob *= self.ADVERSE_ACTION_MULTIPLIER.get(action, 1.0)
        adverse_prob += age_factor * 0.05 # Older patients have higher risk
        adverse_prob += comorbidity_factor * 0.05 # Patients with more comorbidities have higher risk
        adverse_prob += current_severity * 0.02 # Higher severity patients might be more fragile

        if self.rng.rand() < adverse_prob: # An adverse event occurred
            current_severity = min(1.0, current_severity + self.ADVERSE_EFFECT_SEVERITY_INCREASE)
            adverse_event_penalty = -0.5 # Immediate and significant penalty

        self.state[0] = current_severity
        self.t += 1

        # --- 6. Determine done status ---
        done = False
        if self.t >= self.max_t: # Episode ends due to time limit
            done = True
        elif current_severity >= 0.95: # Episode ends due to critical failure
            done = True
        elif current_severity <= 0.05: # Episode ends due to successful remission
            done = True

        # --- 7. Finer Clinical Rewards (QALYs approximated) ---
        # Reward components:
        # a. QALY-like reward: (1 - severity) scaled per step (higher for lower severity)
        reward_qaly_approx = (1.0 - current_severity) * 0.1 # Max 0.1 for perfect health, min 0 for max severity

        # b. Time/Cost penalty: small penalty for each step to encourage efficiency
        reward_time_cost = -0.01

        # Combine step-wise rewards
        reward_step = reward_qaly_approx + reward_time_cost + adverse_event_penalty

        # c. Terminal rewards: significant rewards/penalties for episode outcome
        if done:
            if current_severity <= 0.05: # Success
                reward_step += 1.0
            elif current_severity >= 0.95: # Failure
                reward_step -= 1.0
            # If done by max_t and not success/failure, no additional terminal bonus/penalty here.
            # Could add a continuous terminal reward based on final severity (e.g., 0.5 - current_severity)

        return reward_step, done

    def state_representation(self):
        grid_h, grid_w = 6, 7 # Keeping original grid size for CNN input compatibility

        # Create channels for different state features
        severity_channel = np.full((grid_h, grid_w), self.state[0], dtype=np.float32)
        noise_channel = np.full((grid_h, grid_w), self.state[1], dtype=np.float32)

        # Normalized age and comorbidities as static channels for the patient
        age_normalized = (self.age - 18) / 72.0 # Normalize age 18-90 to 0-1
        comorbidities_normalized = self.comorbidities / 5.0 # Normalize 0-5 comorbidities to 0-1
        age_channel = np.full((grid_h, grid_w), age_normalized, dtype=np.float32)
        comorbidities_channel = np.full((grid_h, grid_w), comorbidities_normalized, dtype=np.float32)

        # Channel for latent effects: Sum of remaining treatment effect magnitudes
        # This provides a dynamic view of pending treatment impacts
        latent_effect_sum = sum(p[2] * (p[1] / (self.LATENCY_DURATION_A if p[0] == 1 else self.LATENCY_DURATION_B)) for p in self.pending_effects) # weighted by remaining duration fraction
        latent_effect_channel = np.full((grid_h, grid_w), latent_effect_sum, dtype=np.float32)

        # Stack all channels to form the state representation
        return np.stack([severity_channel, noise_channel, age_channel, comorbidities_channel, latent_effect_channel])


# ============= Updated AZ-style Neural Network =============
# This AZNet is updated to accept the new number of input channels from ModifiedToyMedicalEnv
class AZNet(nn.Module):
    def __init__(self, in_channels=5, action_size=3, grid_h=6, grid_w=7): # in_channels updated to 5, action_size to 3
        super().__init__()
        self.grid_h = grid_h
        self.grid_w = grid_w
        self.conv = nn.Sequential(
            nn.Conv2d(in_channels, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, padding=1),
            nn.ReLU()
        )
        self.policy_conv = nn.Conv2d(64, 2, kernel_size=1)
        self.policy_fc = nn.Linear(2 * self.grid_h * self.grid_w, action_size)
        self.value_conv = nn.Conv2d(64, 1, kernel_size=1)
        self.value_fc1 = nn.Linear(self.grid_h * self.grid_w, 64)
        self.value_fc2 = nn.Linear(64, 1)

    def forward(self, state_batch):
        # Accept numpy arrays or torch tensors
        if isinstance(state_batch, np.ndarray):
            state_batch = torch.tensor(state_batch, dtype=torch.float32)
        # Ensure batch dimension is present
        if state_batch.dim() == 3:
            state_batch = state_batch.unsqueeze(0)
        state_batch = state_batch.to(DEVICE)

        x = self.conv(state_batch)

        p = F.relu(self.policy_conv(x))
        p = p.view(p.size(0), -1) # Flatten for FC layer
        p = self.policy_fc(p) # Policy logits

        v = F.relu(self.value_conv(x))
        v = v.view(v.size(0), -1) # Flatten for FC layer
        v = F.relu(self.value_fc1(v))
        v = torch.tanh(self.value_fc2(v)).squeeze(-1) # Value output (tanh for [-1, 1] range)
        return p, v

# ============= MCTS Generic =============
class MCTSNode:
    def __init__(self, prior=0.0):
        self.N = 0
        self.W = 0.0
        self.Q = 0.0
        self.P = prior
        self.children = {}

class MCTS:
    def __init__(self, net, action_size, sims=50, cpuct=1.4):
        self.net = net
        self.sims = sims
        self.cpuct = cpuct
        self.action_size = action_size

    def run(self, env):
        root = MCTSNode()
        self._expand(root, env)
        for _ in range(self.sims):
            env_copy = env.clone()
            self._simulate(env_copy, root)
        return root

    def _simulate(self, env, node):
        if not node.children:
            return self._evaluate_and_expand(env, node)

        # selection
        best_a, best_child = None, None
        best_score = -1e9
        for a, ch in node.children.items():
            u = self.cpuct * ch.P * math.sqrt(node.N + 1) / (1 + ch.N)
            score = ch.Q + u
            if score > best_score:
                best_score = score
                best_a = a
                best_child = ch
        reward, done = env.step(best_a)
        if done:
            # Reward is already the terminal reward for AlphaZero's value target
            v = reward
        else:
            v = self._simulate(env, best_child)
        # backup
        best_child.N += 1
        best_child.W += v
        best_child.Q = best_child.W / best_child.N
        node.N += 1
        node.W += v
        node.Q = node.W / node.N
        return v

    def _evaluate_and_expand(self, env, node):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]
            v = float(value.cpu().numpy()[0])

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability (shouldn't happen with softmax on valid actions)
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))
        return v

    def _expand(self, node, env):
        state = env.state_representation()
        state_t = torch.tensor(state, dtype=torch.float32).unsqueeze(0).to(DEVICE)
        with torch.no_grad():
            logits, value = self.net(state_t)
            probs = F.softmax(logits, dim=-1).cpu().numpy()[0]

        valid_actions = env.valid_actions()
        # Distribute probability mass over valid actions only
        total_valid_prob = sum(probs[a] for a in valid_actions)
        if total_valid_prob > 0:
            normalized_probs = {a: probs[a] / total_valid_prob for a in valid_actions}
        else:
            # Fallback if no valid actions have non-zero probability
            normalized_probs = {a: 1.0 / len(valid_actions) for a in valid_actions}

        for a in valid_actions:
            node.children[a] = MCTSNode(prior=normalized_probs.get(a, 0.0))

# ============= Self-play dataset generation =============
def generate_self_play_data(net, scenarios_df, n_games=16, sims=50):
    net.eval()
    dataset = []
    scenario_indices = list(range(len(scenarios_df)))

    for i in tqdm(range(n_games), desc="Generating self-play data"):
        scenario_row = scenarios_df.iloc[scenario_indices[i % len(scenarios_df)]]

        env = ModifiedToyMedicalEnv(
            seed=int(scenario_row['rng_seed']),
            initial_severity=float(scenario_row['initial_severity']),
            age=int(scenario_row['age']),
            comorbidities=int(scenario_row['comorbidities'])
        )

        mcts = MCTS(net, action_size=env.n_actions, sims=sims)

        states, mcts_probs, rewards_history = [], [], [] # rewards_history to store step-wise rewards
        while True:
            root = mcts.run(env)

            counts = np.array([root.children[a].N if a in root.children else 0 for a in range(env.n_actions)], dtype=np.float32)
            if counts.sum() == 0:
                probs = np.zeros_like(counts)
                avail = env.valid_actions()
                if avail:
                    probs[avail] = 1.0 / len(avail)
            else:
                probs = counts / (counts.sum() + 1e-8)

            states.append(env.state_representation())
            mcts_probs.append(probs)

            if probs.sum() == 0 and env.valid_actions():
                action = random.choice(env.valid_actions())
            else:
                action = np.random.choice(np.arange(len(probs)), p=probs)

            reward, done = env.step(action)
            rewards_history.append(reward)

            if done:
                # For AlphaZero, the value target for each state in the trajectory is the final outcome of the episode.
                # Sum of rewards in ModifiedToyMedicalEnv is the episode return.
                episode_return = sum(rewards_history)
                for s, p_dist in zip(states, mcts_probs):
                    dataset.append((s, p_dist, episode_return)) # All states get the final episode return as value target
                break
    return dataset

# ============= Training loop =============
def train_network(net, dataset, epochs=5, batch_size=32, lr=1e-3):
    net.train()
    opt = torch.optim.Adam(net.parameters(), lr=lr)
    for e in range(epochs):
        random.shuffle(dataset)
        losses = []
        for i in range(0, len(dataset), batch_size):
            batch = dataset[i:i+batch_size]
            if not batch:
                continue # Skip empty batches
            states = np.stack([b[0] for b in batch])
            target_p = np.stack([b[1] for b in batch])
            target_v = np.array([b[2] for b in batch], dtype=np.float32)

            states_t = torch.tensor(states, dtype=torch.float32).to(DEVICE)
            target_p_t = torch.tensor(target_p, dtype=torch.float32).to(DEVICE)
            target_v_t = torch.tensor(target_v, dtype=torch.float32).to(DEVICE)

            logits, values = net(states_t)
            loss_v = F.mse_loss(values, target_v_t)
            loss_p = -torch.mean(torch.sum(target_p_t * F.log_softmax(logits, dim=-1), dim=1))
            loss = loss_v + loss_p

            opt.zero_grad()
            loss.backward()
            opt.step()
            losses.append(loss.item())
        print(f'Epoch {e+1}/{epochs} - loss {np.mean(losses):.4f}')

# ============= Evaluate one episode =============
def evaluate_one(net, env_callable, sims=40):
    net.eval()
    env = env_callable()
    mcts = MCTS(net, action_size=env.n_actions, sims=sims)
    traj = []
    episode_rewards = []
    while True:
        root = mcts.run(env)
        counts = np.array([root.children[a].N if a in root.children else 0 for a in range(getattr(env, 'n_actions'))], dtype=np.float32)
        if counts.sum() == 0:
            probs = np.zeros_like(counts)
            avail = env.valid_actions()
            probs[avail] = 1.0 / len(avail)
        else:
            probs = counts / (counts.sum() + 1e-8)
        action = probs.argmax()
        traj.append((env.state_representation(), probs, action))
        reward, done = env.step(action)
        episode_rewards.append(reward)
        if done:
            return sum(episode_rewards), traj


# --- Main execution logic with refined hyperparameters ---
set_seed(123)

# Explicitly load df_scenarios here to ensure it's defined.
# This assumes 'synthetic_clinical_scenarios.csv' exists in the current environment.
try:
    df_scenarios = pd.read_csv('synthetic_clinical_scenarios.csv')
    print(f"Loaded {len(df_scenarios)} synthetic clinical scenarios from 'synthetic_clinical_scenarios.csv'.")
except FileNotFoundError:
    print("Error: 'synthetic_clinical_scenarios.csv' not found. Creating a dummy DataFrame for demonstration.")
    # Fallback if df_scenarios isn't preloaded and file not found, create a dummy one
    data = {
    'patient_id': np.arange(1, 51),
    'age': np.random.randint(18, 91, 50),
    'comorbidities': np.random.randint(0, 6, 50),
    'initial_severity': np.random.uniform(0.2, 0.8, 50),
    'rng_seed': np.arange(50, 100)
    }
    df_scenarios = pd.DataFrame(data)
    print("Using dummy df_scenarios for demonstration.")


# Self-play data generation parameters
NEW_N_GAMES = 50 # Increased from 16 to 50
NEW_MCTS_SIMS = 75 # Adjusted from 50 to 75

# Training network parameters
NEW_EPOCHS = 10 # Increased from 6 to 10
NEW_BATCH_SIZE = 32 # Increased from 16 to 32
NEW_LR = 5e-4 # Adjusted from 1e-3 to 5e-4

print(f"New Hyperparameters:\n  n_games: {NEW_N_GAMES}\n  sims: {NEW_MCTS_SIMS}\n  epochs: {NEW_EPOCHS}\n  batch_size: {NEW_BATCH_SIZE}\n  lr: {NEW_LR}")

# Create a dummy instance of ModifiedToyMedicalEnv to get state shape and action size
dummy_env = ModifiedToyMedicalEnv(seed=0, age=50, comorbidities=1, initial_severity=0.5)
input_channels = dummy_env.state_representation().shape[0]
action_size = dummy_env.n_actions

# Initialize a new AZNet with the correct dimensions
net_refined = AZNet(in_channels=input_channels, action_size=action_size,
                    grid_h=6, grid_w=7).to(DEVICE)

print('\nStarting self-play data generation with refined parameters...')
# Call the modified generate_self_play_data function with df_scenarios and new params
dataset_refined = generate_self_play_data(net_refined, df_scenarios,
                                            n_games=NEW_N_GAMES, sims=NEW_MCTS_SIMS)
print('Refined Dataset size:', len(dataset_refined))

if len(dataset_refined) == 0:
    print('Warning: Refined dataset is empty. Reduce sims or increase n_games.')
else:
    print('\nStarting network training with refined parameters...')
    train_network(net_refined, dataset_refined, epochs=NEW_EPOCHS,
                  batch_size=NEW_BATCH_SIZE, lr=NEW_LR)
    save_model(net_refined, 'aznet_med_colab_refined.pth')
    print('Saved refined model to aznet_med_colab_refined.pth')

# Evaluate one episode with the refined environment and network
print('\nEvaluating one episode with refined network...')
# Use a different seed for evaluation to ensure varied scenarios
eval_env_args = {'age': 65, 'comorbidities': 2, 'initial_severity': 0.6, 'seed': 99} # Changed 'rng_seed' to 'seed'
r_refined, traj_refined = evaluate_one(net_refined,
                                         lambda: ModifiedToyMedicalEnv(**eval_env_args),
                                         sims=NEW_MCTS_SIMS)
print('Episode finished with refined network, reward:', r_refined)
````
